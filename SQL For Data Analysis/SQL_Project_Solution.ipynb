# 🗺️ 2. Database Understanding: AdventureWorks Snowflake Schema

Before diving into the data, it's crucial to understand the database structure. AdventureWorks uses a **Snowflake Schema**, which is an extension of the Star Schema. It normalizes dimension tables into multiple related tables, reducing data redundancy but potentially increasing join complexity.

### ❄️ Snowflake Schema Overview

Imagine a central fact table (like `FactInternetSales`) surrounded by primary dimension tables, which themselves are linked to further sub-dimension tables. This design helps organize and store data efficiently, especially for analytical reporting.

### 📊 Key Tables in Our Analysis

Here are the primary tables we'll be working with and their purposes:

*   **`FactInternetSales`**: This is our central **Fact Table**. It contains transactional sales data, such as order quantity, sales amount, and foreign keys linking to various dimension tables. Each row represents a single line item of a sale.

*   **`DimCustomer`**: A **Dimension Table** containing detailed information about each customer, including their name, demographics, geographic location, and contact information.

*   **`DimProduct`**: A **Dimension Table** providing attributes for each product, such as product name, color, size, and standard cost. It links to `DimProductSubcategory`.

*   **`DimProductSubcategory`**: A **Dimension Table** that categorizes products into subcategories (e.g., 'Mountain Bikes', 'Road Bikes'). It links to `DimProductCategory`.

*   **`DimProductCategory`**: A **Dimension Table** providing high-level categories for products (e.g., 'Bikes', 'Components', 'Clothing').

*   **`DimDate`**: A **Dimension Table** containing various date-related attributes (e.g., year, month, day, day of week) that can be used for time-based analysis.

*   **`DimSalesTerritory`**: A **Dimension Table** describing the sales regions, including territory name, country, and group.

## Level 1: Foundations (Easy)

These questions focus on basic `SELECT` statements, filtering (`WHERE`), sorting (`ORDER BY`), and understanding distinct values. These are your building blocks!

### Question 1: Explore All Products

**Business Explanation**: Management wants to see a complete list of all products AdventureWorks sells to get an initial overview of their catalog.

**Hint**: A simple `SELECT *` on the product dimension table will do the trick.

```sql
SELECT * FROM DimProduct;
```

**Expected Output**: A table showing all columns for every product in the `DimProduct` table.

**Business Insight**: Provides a foundational understanding of the available product data, including product names, colors, sizes, and costs.

### Question 2: Find Customers in Specific Regions

**Business Explanation**: The marketing team is planning a targeted campaign for customers located in 'Australia' or 'Canada'. They need a list of these customers.

**Hint**: Use `WHERE` with `OR` or `IN` on the `EnglishCountryRegionName` column in `DimCustomer`.

```sql
SELECT CustomerKey, FirstName, LastName, EnglishCountryRegionName
FROM DimCustomer
WHERE EnglishCountryRegionName IN ('Australia', 'Canada');
```

**Expected Output**: A table listing customers (e.g., `CustomerKey`, `FirstName`, `LastName`, `EnglishCountryRegionName`) who reside in Australia or Canada.

**Business Insight**: Helps in segmenting customers by geography for targeted marketing or regional sales analysis.

### Question 3: Top 10 Latest Sales Transactions

**Business Explanation**: Sales managers want to review the most recent sales transactions to keep an eye on current sales activities.

**Hint**: Select from `FactInternetSales`, order by a date column in descending order, and limit the results.

```sql
SELECT TOP 10 *
FROM FactInternetSales
ORDER BY OrderDate DESC;
```

**Expected Output**: The top 10 rows from the `FactInternetSales` table, ordered by `OrderDate` in descending order.

**Business Insight**: Provides a quick snapshot of recent sales, which can help in identifying immediate trends or issues.

### Question 4: Unique Product Colors Available

**Business Explanation**: The product development team wants to understand the variety of colors available across all products. They need a list of all unique colors.

**Hint**: Use `DISTINCT` on the `Color` column in `DimProduct`.

### Question 5: Products Priced Above a Certain Threshold

**Business Explanation**: Management wants to identify premium products. They define premium as any product with a `ListPrice` greater than $1000.

**Hint**: Filter `DimProduct` using `WHERE` on the `ListPrice` column.

## Level 2: Aggregations (Intermediate)

These questions involve using aggregate functions (`COUNT`, `SUM`, `AVG`, `MIN`, `MAX`) with `GROUP BY` and `HAVING` to summarize data and find patterns.

### Question 6: Total Sales Quantity by Product Category

**Business Explanation**: The sales department wants to know which product categories are selling the most units.

**Hint**: You'll need to join `FactInternetSales` with `DimProduct` and `DimProductCategory` to get the category name, then `GROUP BY` category and `SUM` the `OrderQuantity`.

```sql
SELECT
    dpc.EnglishProductCategoryName AS ProductCategoryName,
    SUM(fis.OrderQuantity) AS TotalQuantitySold
FROM FactInternetSales AS fis
JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
JOIN DimProductSubcategory AS dpsc ON dp.ProductSubcategoryKey = dpsc.ProductSubcategoryKey
JOIN DimProductCategory AS dpc ON dpsc.ProductCategoryKey = dpc.ProductCategoryKey
GROUP BY dpc.EnglishProductCategoryName
ORDER BY TotalQuantitySold DESC;
```

**Expected Output**: Two columns: `ProductCategoryName` and `TotalQuantitySold`, showing the total number of units sold for each category.

**Business Insight**: Helps identify the most popular product categories, guiding inventory and marketing focus.

### Question 7: Number of Customers per Sales Territory

**Business Explanation**: The sales operations team needs to understand the customer distribution across different sales territories to allocate resources effectively.

**Hint**: Join `DimCustomer` with `DimSalesTerritory`, then `GROUP BY` territory name and `COUNT` distinct customers.

```sql
SELECT
    dst.SalesTerritoryRegion AS SalesTerritoryRegion,
    COUNT(DISTINCT dc.CustomerKey) AS NumberOfCustomers
FROM DimCustomer AS dc
JOIN DimSalesTerritory AS dst ON dc.SalesTerritoryKey = dst.SalesTerritoryKey
GROUP BY dst.SalesTerritoryRegion
ORDER BY NumberOfCustomers DESC;
```

**Expected Output**: Two columns: `SalesTerritoryRegion` and `NumberOfCustomers`, showing the count of unique customers in each region.

**Business Insight**: Informs sales team staffing and regional marketing budget allocation.

### Question 8: Categories with Total Sales Revenue Exceeding a Threshold

**Business Explanation**: The finance department wants to highlight product categories that generate significant revenue, specifically those with total sales greater than $500,000.

**Hint**: Similar to Q6, but use `SUM(SalesAmount)` and filter with `HAVING`.

```sql
SELECT
    dpc.EnglishProductCategoryName AS ProductCategoryName,
    SUM(fis.SalesAmount) AS TotalSalesRevenue
FROM FactInternetSales AS fis
JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
JOIN DimProductSubcategory AS dpsc ON dp.ProductSubcategoryKey = dpsc.ProductSubcategoryKey
JOIN DimProductCategory AS dpc ON dpsc.ProductCategoryKey = dpc.ProductCategoryKey
GROUP BY dpc.EnglishProductCategoryName
HAVING SUM(fis.SalesAmount) > 500000
ORDER BY TotalSalesRevenue DESC;
```

**Expected Output**: `ProductCategoryName` and `TotalSalesRevenue` for categories where the total sales amount is greater than $500,000.

**Business Insight**: Pinpoints high-performing categories, which can influence investment decisions and strategic focus.

## Level 3: Joining the Dots (Advanced Intermediate)

These questions require combining data from multiple tables using `JOIN` clauses, navigating the snowflake schema to enrich our analysis.

### Question 9: Total Revenue by Product Subcategory

**Business Explanation**: A more granular view of product performance is needed. The product team wants to know the total revenue generated by each product subcategory.

**Hint**: You'll need `FactInternetSales`, `DimProduct`, and `DimProductSubcategory`. Join them and `SUM` `SalesAmount`, then `GROUP BY` subcategory name.

```sql
SELECT
    dpsc.EnglishProductSubcategoryName AS ProductSubcategoryName,
    SUM(fis.SalesAmount) AS TotalRevenue
FROM FactInternetSales AS fis
JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
JOIN DimProductSubcategory AS dpsc ON dp.ProductSubcategoryKey = dpsc.ProductSubcategoryKey
GROUP BY dpsc.EnglishProductSubcategoryName
ORDER BY TotalRevenue DESC;
```

**Expected Output**: `ProductSubcategoryName` and `TotalRevenue` for each subcategory.

**Business Insight**: Offers a detailed understanding of which specific product lines are driving revenue, helping with product portfolio management.

### Question 10: Top 5 Customers by Total Sales Amount

**Business Explanation**: The loyalty program manager wants to identify the most valuable customers based on their total purchase value.

**Hint**: Join `FactInternetSales` with `DimCustomer`. `SUM(SalesAmount)` grouped by customer, then `ORDER BY` total sales and `TOP` 5.

```sql
SELECT TOP 5
    dc.FirstName,
    dc.LastName,
    SUM(fis.SalesAmount) AS TotalSalesAmount
FROM FactInternetSales AS fis
JOIN DimCustomer AS dc ON fis.CustomerKey = dc.CustomerKey
GROUP BY dc.CustomerKey, dc.FirstName, dc.LastName
ORDER BY TotalSalesAmount DESC;
```

**Expected Output**: `FirstName`, `LastName`, and `TotalSalesAmount` for the top 5 customers.

**Business Insight**: Helps identify VIP customers for exclusive promotions, personalized services, or loyalty programs.

## Level 4: Functions & Logic (Advanced)

These questions involve applying SQL functions (string, date, numeric) and conditional logic (`CASE` statements) to transform and analyze data more deeply.

### Question 11: Sales Performance by Quarter and Year

**Business Explanation**: Finance needs to see sales performance broken down by fiscal quarter and year for quarterly reporting.

**Hint**: Use `YEAR` and `MONTH` from `DimDate` to derive the quarter or use a dedicated `CalendarQuarter` column if available, then `SUM(SalesAmount)` grouped by year and quarter.

```sql
SELECT
    dd.CalendarYear,
    dd.CalendarQuarter,
    SUM(fis.SalesAmount) AS TotalSalesAmount
FROM FactInternetSales AS fis
JOIN DimDate AS dd ON fis.OrderDateKey = dd.DateKey
GROUP BY dd.CalendarYear, dd.CalendarQuarter
ORDER BY dd.CalendarYear, dd.CalendarQuarter;
```

**Expected Output**: `CalendarYear`, `CalendarQuarter`, and `TotalSalesAmount` for each quarter of each year.

**Business Insight**: Provides a structured view of sales trends over time, crucial for financial planning and performance review.

### Question 12: Product Price Categories

**Business Explanation**: Management wants to classify products into 'Economy', 'Standard', and 'Premium' based on their `ListPrice` for pricing strategy analysis.

**Hint**: Use a `CASE` statement on `ListPrice` from `DimProduct` to assign categories. Define your own price ranges.

```sql
SELECT
    EnglishProductName,
    ListPrice,
    CASE
        WHEN ListPrice < 500 THEN 'Economy'
        WHEN ListPrice >= 500 AND ListPrice < 1500 THEN 'Standard'
        ELSE 'Premium'
    END AS PriceCategory
FROM DimProduct
ORDER BY ListPrice DESC;
```

**Expected Output**: `EnglishProductName`, `ListPrice`, and a new column `PriceCategory` (e.g., 'Economy', 'Standard', 'Premium') for each product.

**Business Insight**: Categorizes products for targeted marketing, pricing strategies, and understanding product mix by value.

## Level 5: Advanced Analytics (Challenging)

These questions involve more complex SQL constructs like subqueries, Common Table Expressions (CTEs), and window functions, allowing for deeper analytical insights.

### Question 13: Top Selling Product in Each Category

**Business Explanation**: The product development team wants to identify the single best-selling product (by total sales quantity) within *each* product category.

**Hint**: Use a subquery or CTE to calculate total sales quantity per product. Then, use `RANK()` with `PARTITION BY ProductCategoryName` and `ORDER BY TotalQuantitySold DESC`.

```sql
WITH ProductSales AS (
    SELECT
        dpc.EnglishProductCategoryName AS ProductCategoryName,
        dp.EnglishProductName,
        SUM(fis.OrderQuantity) AS TotalQuantitySold,
        ROW_NUMBER() OVER(PARTITION BY dpc.EnglishProductCategoryName ORDER BY SUM(fis.OrderQuantity) DESC) AS rn
    FROM FactInternetSales AS fis
    JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
    JOIN DimProductSubcategory AS dpsc ON dp.ProductSubcategoryKey = dpsc.ProductSubcategoryKey
    JOIN DimProductCategory AS dpc ON dpsc.ProductCategoryKey = dpc.ProductCategoryKey
    GROUP BY dpc.EnglishProductCategoryName, dp.EnglishProductName
)
SELECT
    ProductCategoryName,
    EnglishProductName,
    TotalQuantitySold
FROM ProductSales
WHERE rn = 1
ORDER BY ProductCategoryName;
```

**Expected Output**: `ProductCategoryName`, `EnglishProductName`, and `TotalQuantitySold` for the top product in each category.

**Business Insight**: Helps pinpoint star products within each segment, informing product development, marketing, and inventory decisions.
