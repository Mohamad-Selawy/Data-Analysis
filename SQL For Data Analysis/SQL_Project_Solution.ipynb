# 🗺️ 2. Database Understanding: AdventureWorks Snowflake Schema

Before diving into the data, it's crucial to understand the database structure. AdventureWorks uses a **Snowflake Schema**, which is an extension of the Star Schema. It normalizes dimension tables into multiple related tables, reducing data redundancy but potentially increasing join complexity.

### ❄️ Snowflake Schema Overview

Imagine a central fact table (like `FactInternetSales`) surrounded by primary dimension tables, which themselves are linked to further sub-dimension tables. This design helps organize and store data efficiently, especially for analytical reporting.

### 📊 Key Tables in Our Analysis

Here are the primary tables we'll be working with and their purposes:

*   **`FactInternetSales`**: This is our central **Fact Table**. It contains transactional sales data, such as order quantity, sales amount, and foreign keys linking to various dimension tables. Each row represents a single line item of a sale.

*   **`DimCustomer`**: A **Dimension Table** containing detailed information about each customer, including their name, demographics, geographic location, and contact information.

*   **`DimProduct`**: A **Dimension Table** providing attributes for each product, such as product name, color, size, and standard cost. It links to `DimProductSubcategory`.

*   **`DimProductSubcategory`**: A **Dimension Table** that categorizes products into subcategories (e.g., 'Mountain Bikes', 'Road Bikes'). It links to `DimProductCategory`.

*   **`DimProductCategory`**: A **Dimension Table** providing high-level categories for products (e.g., 'Bikes', 'Components', 'Clothing').

*   **`DimDate`**: A **Dimension Table** containing various date-related attributes (e.g., year, month, day, day of week) that can be used for time-based analysis.

*   **`DimSalesTerritory`**: A **Dimension Table** describing the sales regions, including territory name, country, and group.

### Question 1: Explore All Products

**Hint**: A simple `SELECT *` on the product dimension table will do the trick.

### Question 2: Find Customers in Specific Regions

**Hint**: Use `WHERE` with `OR` or `IN` on the `EnglishCountryRegionName` column in `DimCustomer`.

## Level 2: Aggregations (Intermediate)

These questions involve using aggregate functions (`COUNT`, `SUM`, `AVG`, `MIN`, `MAX`) with `GROUP BY` and `HAVING` to summarize data and find patterns.

### Question 6: Total Sales Quantity by Product Category

**Business Explanation**: The sales department wants to know which product categories are selling the most units.

**Hint**: You'll need to join `FactInternetSales` with `DimProduct` and `DimProductCategory` to get the category name, then `GROUP BY` category and `SUM` the `OrderQuantity`.

```sql
SELECT
    dpc.EnglishProductCategoryName AS ProductCategoryName,
    SUM(fis.OrderQuantity) AS TotalQuantitySold
FROM FactInternetSales AS fis
JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
JOIN DimProductSubcategory AS dpsc ON dp.ProductSubcategoryKey = dpsc.ProductSubcategoryKey
JOIN DimProductCategory AS dpc ON dpsc.ProductCategoryKey = dpc.ProductCategoryKey
GROUP BY dpc.EnglishProductCategoryName
ORDER BY TotalQuantitySold DESC;
```

**Expected Output**: Two columns: `ProductCategoryName` and `TotalQuantitySold`, showing the total number of units sold for each category.

**Business Insight**: Helps identify the most popular product categories, guiding inventory and marketing focus.

### Question 7: Number of Customers per Sales Territory

**Business Explanation**: The sales operations team needs to understand the customer distribution across different sales territories to allocate resources effectively.

**Hint**: Join `DimCustomer` with `DimSalesTerritory`, then `GROUP BY` territory name and `COUNT` distinct customers.

```sql
SELECT
    dst.SalesTerritoryRegion AS SalesTerritoryRegion,
    COUNT(DISTINCT dc.CustomerKey) AS NumberOfCustomers
FROM DimCustomer AS dc
JOIN DimSalesTerritory AS dst ON dc.SalesTerritoryKey = dst.SalesTerritoryKey
GROUP BY dst.SalesTerritoryRegion
ORDER BY NumberOfCustomers DESC;
```

**Expected Output**: Two columns: `SalesTerritoryRegion` and `NumberOfCustomers`, showing the count of unique customers in each region.

**Business Insight**: Informs sales team staffing and regional marketing budget allocation.

### Question 8: Average Sales Amount per Order

**Business Explanation**: Management wants to know the average value of each sales order to understand typical customer spending habits.

**Hint**: Use `AVG()` on `SalesAmount` and `GROUP BY` a unique order identifier, or use `SalesOrderNumber` in `FactInternetSales` if orders can have multiple line items.

```sql
SELECT
    SalesOrderNumber,
    AVG(SalesAmount) AS AverageSalesAmountPerOrder
FROM FactInternetSales
GROUP BY SalesOrderNumber
ORDER BY SalesOrderNumber;
```

**Expected Output**: A single value representing the average `SalesAmount` per order number.

**Business Insight**: Provides a benchmark for average transaction size, useful for setting sales targets or assessing promotional impact.

### Question 9: Categories with Total Sales Revenue Exceeding a Threshold

**Business Explanation**: The finance department wants to highlight product categories that generate significant revenue, specifically those with total sales greater than $500,000.

**Hint**: Similar to Q6, but use `SUM(SalesAmount)` and filter with `HAVING`.

```sql
SELECT
    dpc.EnglishProductCategoryName AS ProductCategoryName,
    SUM(fis.SalesAmount) AS TotalSalesRevenue
FROM FactInternetSales AS fis
JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
JOIN DimProductSubcategory AS dpsc ON dp.ProductSubcategoryKey = dpsc.ProductSubcategoryKey
JOIN DimProductCategory AS dpc ON dpsc.ProductCategoryKey = dpc.ProductCategoryKey
GROUP BY dpc.EnglishProductCategoryName
HAVING SUM(fis.SalesAmount) > 500000
ORDER BY TotalSalesRevenue DESC;
```

**Expected Output**: `ProductCategoryName` and `TotalSalesRevenue` for categories where the total sales amount is greater than $500,000.

**Business Insight**: Pinpoints high-performing categories, which can influence investment decisions and strategic focus.

## Level 3: Joining the Dots (Advanced Intermediate)

These questions require combining data from multiple tables using `JOIN` clauses, navigating the snowflake schema to enrich our analysis.

### Question 10: Total Revenue by Product Subcategory

**Business Explanation**: A more granular view of product performance is needed. The product team wants to know the total revenue generated by each product subcategory.

**Hint**: You'll need `FactInternetSales`, `DimProduct`, and `DimProductSubcategory`. Join them and `SUM` `SalesAmount`, then `GROUP BY` subcategory name.

```sql
SELECT
    dpsc.EnglishProductSubcategoryName AS ProductSubcategoryName,
    SUM(fis.SalesAmount) AS TotalRevenue
FROM FactInternetSales AS fis
JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
JOIN DimProductSubcategory AS dpsc ON dp.ProductSubcategoryKey = dpsc.ProductSubcategoryKey
GROUP BY dpsc.EnglishProductSubcategoryName
ORDER BY TotalRevenue DESC;
```

**Expected Output**: `ProductSubcategoryName` and `TotalRevenue` for each subcategory.

**Business Insight**: Offers a detailed understanding of which specific product lines are driving revenue, helping with product portfolio management.

### Question 11: Customer Sales by Country

**Business Explanation**: The international sales team wants to analyze customer purchasing behavior aggregated by their country of residence.

**Hint**: Join `FactInternetSales` with `DimCustomer` to get customer information, and then `DimSalesTerritory` to get country details. `GROUP BY` country and `SUM` `SalesAmount`.

```sql
SELECT
    dst.EnglishCountryRegionName AS EnglishCountryRegionName,
    SUM(fis.SalesAmount) AS TotalSalesRevenue
FROM FactInternetSales AS fis
JOIN DimCustomer AS dc ON fis.CustomerKey = dc.CustomerKey
JOIN DimSalesTerritory AS dst ON dc.SalesTerritoryKey = dst.SalesTerritoryKey
GROUP BY dst.EnglishCountryRegionName
ORDER BY TotalSalesRevenue DESC;
```

**Expected Output**: `EnglishCountryRegionName` and `TotalSalesRevenue` by country.

**Business Insight**: Reveals geographical sales performance, highlighting strong and weak markets for strategic focus.

### Question 12: Products Sold in 'Europe' Sales Territory

**Business Explanation**: The European sales director wants a list of all products that have been sold in the 'Europe' sales territory.

**Hint**: You'll need to join `FactInternetSales` with `DimProduct` and `DimSalesTerritory`. Filter for the 'Europe' territory and use `DISTINCT` product names.

```sql
SELECT DISTINCT
    dp.EnglishProductName
FROM FactInternetSales AS fis
JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
JOIN DimCustomer AS dc ON fis.CustomerKey = dc.CustomerKey
JOIN DimSalesTerritory AS dst ON dc.SalesTerritoryKey = dst.SalesTerritoryKey
WHERE dst.SalesTerritoryGroup = 'Europe'
ORDER BY dp.EnglishProductName;
```

**Expected Output**: A distinct list of `EnglishProductName` that were sold in the 'Europe' territory.

**Business Insight**: Identifies products popular or available in specific regions, useful for localized marketing and inventory planning.

### Question 13: Top 5 Customers by Total Sales Amount

**Business Explanation**: The loyalty program manager wants to identify the most valuable customers based on their total purchase value.

**Hint**: Join `FactInternetSales` with `DimCustomer`. `SUM(SalesAmount)` grouped by customer, then `ORDER BY` total sales and `TOP` 5.

```sql
SELECT TOP 5
    dc.FirstName,
    dc.LastName,
    SUM(fis.SalesAmount) AS TotalSalesAmount
FROM FactInternetSales AS fis
JOIN DimCustomer AS dc ON fis.CustomerKey = dc.CustomerKey
GROUP BY dc.CustomerKey, dc.FirstName, dc.LastName
ORDER BY TotalSalesAmount DESC;
```

**Expected Output**: `FirstName`, `LastName`, and `TotalSalesAmount` for the top 5 customers.

**Business Insight**: Helps identify VIP customers for exclusive promotions, personalized services, or loyalty programs.

## Level 4: Functions & Logic (Advanced)

These questions involve applying SQL functions (string, date, numeric) and conditional logic (`CASE` statements) to transform and analyze data more deeply.

### Question 14: Full Customer Names and Email Domains

**Business Explanation**: The marketing team needs customer full names for personalized communications and wants to analyze email domains for partnership opportunities.

**Hint**: Use `CONCAT` for full name and a string function like `SUBSTRING` or `CHARINDEX` to extract the email domain from `EmailAddress` in `DimCustomer`.

```sql
SELECT
    CONCAT(FirstName, ' ', LastName) AS CustomerFullName,
    SUBSTRING(EmailAddress, CHARINDEX('@', EmailAddress) + 1, LEN(EmailAddress)) AS EmailDomain
FROM DimCustomer
WHERE EmailAddress IS NOT NULL;
```

**Expected Output**: `CustomerFullName` (e.g., 'John Doe') and `EmailDomain` (e.g., 'example.com') for each customer.

**Business Insight**: Useful for personalizing communications and understanding customer email service providers or company affiliations.

### Question 15: Sales Performance by Quarter and Year

**Business Explanation**: Finance needs to see sales performance broken down by fiscal quarter and year for quarterly reporting.

**Hint**: Use `YEAR` and `MONTH` from `DimDate` to derive the quarter or use a dedicated `CalendarQuarter` column if available, then `SUM(SalesAmount)` grouped by year and quarter.

```sql
SELECT
    dd.CalendarYear,
    dd.CalendarQuarter,
    SUM(fis.SalesAmount) AS TotalSalesAmount
FROM FactInternetSales AS fis
JOIN DimDate AS dd ON fis.OrderDateKey = dd.DateKey
GROUP BY dd.CalendarYear, dd.CalendarQuarter
ORDER BY dd.CalendarYear, dd.CalendarQuarter;
```

**Expected Output**: `CalendarYear`, `CalendarQuarter`, and `TotalSalesAmount` for each quarter of each year.

**Business Insight**: Provides a structured view of sales trends over time, crucial for financial planning and performance review.

### Question 16: Product Price Categories

**Business Explanation**: Management wants to classify products into 'Economy', 'Standard', and 'Premium' based on their `ListPrice` for pricing strategy analysis.

**Hint**: Use a `CASE` statement on `ListPrice` from `DimProduct` to assign categories. Define your own price ranges.

```sql
SELECT
    EnglishProductName,
    ListPrice,
    CASE
        WHEN ListPrice < 500 THEN 'Economy'
        WHEN ListPrice >= 500 AND ListPrice < 1500 THEN 'Standard'
        ELSE 'Premium'
    END AS PriceCategory
FROM DimProduct
ORDER BY ListPrice DESC;
```

**Expected Output**: `EnglishProductName`, `ListPrice`, and a new column `PriceCategory` (e.g., 'Economy', 'Standard', 'Premium') for each product.

**Business Insight**: Categorizes products for targeted marketing, pricing strategies, and understanding product mix by value.

## Level 5: Advanced Analytics (Challenging)

These questions involve more complex SQL constructs like subqueries, Common Table Expressions (CTEs), and window functions, allowing for deeper analytical insights.

### Question 17: Top Selling Product in Each Category

**Business Explanation**: The product development team wants to identify the single best-selling product (by total sales quantity) within *each* product category.

**Hint**: Use a subquery or CTE to calculate total sales quantity per product. Then, use `ROW_NUMBER()` or `RANK()` with `PARTITION BY ProductCategoryName` and `ORDER BY TotalQuantitySold DESC`.

```sql
WITH ProductSales AS (
    SELECT
        dpc.EnglishProductCategoryName AS ProductCategoryName,
        dp.EnglishProductName,
        SUM(fis.OrderQuantity) AS TotalQuantitySold,
        ROW_NUMBER() OVER(PARTITION BY dpc.EnglishProductCategoryName ORDER BY SUM(fis.OrderQuantity) DESC) AS rn
    FROM FactInternetSales AS fis
    JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
    JOIN DimProductSubcategory AS dpsc ON dp.ProductSubcategoryKey = dpsc.ProductSubcategoryKey
    JOIN DimProductCategory AS dpc ON dpsc.ProductCategoryKey = dpc.ProductCategoryKey
    GROUP BY dpc.EnglishProductCategoryName, dp.EnglishProductName
)
SELECT
    ProductCategoryName,
    EnglishProductName,
    TotalQuantitySold
FROM ProductSales
WHERE rn = 1
ORDER BY ProductCategoryName;
```

**Expected Output**: `ProductCategoryName`, `EnglishProductName`, and `TotalQuantitySold` for the top product in each category.

**Business Insight**: Helps pinpoint star products within each segment, informing product development, marketing, and inventory decisions.

### Question 18: Running Total of Sales Over Time

**Business Explanation**: The finance team wants to see the cumulative (running) total of `SalesAmount` day by day to track overall business growth.

**Hint**: Use a window function `SUM(SalesAmount) OVER (ORDER BY OrderDate)`.

```sql
SELECT
    DISTINCT OrderDate,
    SUM(SalesAmount) OVER (PARTITION BY OrderDate) AS DailySalesAmount,
    SUM(SalesAmount) OVER (ORDER BY OrderDate) AS RunningTotalSales
FROM FactInternetSales
ORDER BY OrderDate;
```

**Expected Output**: `OrderDate` (distinct), `DailySalesAmount`, and `RunningTotalSales` showing the cumulative sum.

**Business Insight**: Visualizes overall sales growth trajectory, useful for understanding business momentum and identifying periods of rapid increase or stagnation.

### Question 19: Customers Who Purchased More Than Once

**Business Explanation**: The customer retention team wants to identify repeat customers to target them with loyalty incentives.

**Hint**: Use a subquery or CTE to count orders per customer, then filter for counts greater than 1.

```sql
SELECT
    dc.CustomerKey,
    dc.FirstName,
    dc.LastName,
    COUNT(DISTINCT fis.SalesOrderNumber) AS NumberOfOrders
FROM FactInternetSales AS fis
JOIN DimCustomer AS dc ON fis.CustomerKey = dc.CustomerKey
GROUP BY dc.CustomerKey, dc.FirstName, dc.LastName
HAVING COUNT(DISTINCT fis.SalesOrderNumber) > 1
ORDER BY NumberOfOrders DESC;
```

**Expected Output**: `CustomerKey`, `FirstName`, `LastName`, and `NumberOfOrders` for customers who have placed more than one order.

**Business Insight**: Identifies repeat customers, a crucial segment for loyalty programs and personalized marketing to encourage continued engagement.

### Question 20: Monthly Sales Growth Percentage

**Business Explanation**: Senior management wants to understand month-over-month sales growth to assess business expansion and seasonality.

**Hint**: Calculate monthly sales using a CTE. Then use `LAG()` window function to get the previous month's sales, and calculate the percentage difference.

```sql
WITH MonthlySales AS (
    SELECT
        dd.CalendarYear,
        dd.MonthNumberOfYear AS CalendarMonth,
        SUM(fis.SalesAmount) AS MonthlySalesAmount
    FROM FactInternetSales AS fis
    JOIN DimDate AS dd ON fis.OrderDateKey = dd.DateKey
    GROUP BY dd.CalendarYear, dd.MonthNumberOfYear
),
SalesWithPreviousMonth AS (
    SELECT
        CalendarYear,
        CalendarMonth,
        MonthlySalesAmount,
        LAG(MonthlySalesAmount) OVER (PARTITION BY CalendarYear ORDER BY CalendarMonth) AS PreviousMonthSalesAmount
    FROM MonthlySales
)
SELECT
    CalendarYear,
    CalendarMonth,
    MonthlySalesAmount,
    PreviousMonthSalesAmount,
    CASE
        WHEN PreviousMonthSalesAmount = 0 THEN NULL -- Handle division by zero for the first month of the year
        ELSE ((MonthlySalesAmount - PreviousMonthSalesAmount) / PreviousMonthSalesAmount) * 100
    END AS GrowthPercentage
FROM SalesWithPreviousMonth
ORDER BY CalendarYear, CalendarMonth;
```

**Expected Output**: `CalendarYear`, `CalendarMonth`, `MonthlySales`, `PreviousMonthSales`, and `GrowthPercentage`.

**Business Insight**: Provides critical metrics for evaluating business growth, identifying seasonal patterns, and assessing the impact of monthly initiatives.

# 🛠️ 4. Data Quality Assessment

Before relying on our analysis, it's vital to perform data quality checks. Poor data quality can lead to inaccurate insights and flawed business decisions. Here, we'll write SQL queries to identify common data issues.

### 🚫 Why Data Quality Matters

*   **Accuracy**: Ensures our reports and analyses reflect the true state of the business.
*   **Reliability**: Builds trust in the data for decision-makers.
*   **Integrity**: Maintains the consistency and validity of relationships between tables.
*   **Completeness**: Ensures we aren't missing critical information.

---

## 📈 Checks for NULL Values

`NULL` values indicate missing information. Depending on the column, missing data can severely impact aggregations, joins, and filter conditions.

### Check 1: Products with Missing English Product Name

**Business Explanation**: Every product should have a name. Missing names make it impossible to identify or categorize products properly.

**Hint**: Look for `IS NULL` on the `EnglishProductName` column in `DimProduct`.

```sql
SELECT * FROM DimProduct
WHERE EnglishProductName IS NULL;
```

**Expected Output**: Any rows from `DimProduct` where `EnglishProductName` is NULL.

**Business Insight**: Identifies products that are poorly described, making them difficult to track or sell.

### Check 2: Customers with Missing Email Addresses

**Business Explanation**: The marketing team relies on email addresses for customer communication. Missing emails mean lost opportunities for engagement.

**Hint**: Check for `IS NULL` on the `EmailAddress` column in `DimCustomer`.

```sql
SELECT * FROM DimCustomer
WHERE EmailAddress IS NULL;
```

**Expected Output**: Any rows from `DimCustomer` where `EmailAddress` is NULL.

**Business Insight**: Quantifies the reach for email marketing campaigns and highlights areas for data collection improvement.

### Check 3: Sales Records with Missing Sales Amount

**Business Explanation**: Sales transactions without an amount are incomplete and can distort total revenue calculations.

**Hint**: Check for `IS NULL` on `SalesAmount` in `FactInternetSales`.

```sql
SELECT * FROM FactInternetSales
WHERE SalesAmount IS NULL;
```

**Expected Output**: Any rows from `FactInternetSales` where `SalesAmount` is NULL.

**Business Insight**: Ensures financial data accuracy, preventing misreporting of total revenue and average transaction values.

## 🔁 Checks for Duplicate Keys

Primary keys should be unique. Duplicates can lead to miscounts, incorrect joins, and ambiguity in identifying records.

### Check 4: Duplicate Product Keys

**Business Explanation**: Every product should have a unique `ProductKey`. Duplicates could mean data entry errors or issues in the product catalog system.

**Hint**: Use `GROUP BY ProductKey` and `COUNT(*) > 1`.

```sql
SELECT ProductKey, COUNT(*) AS CountOfDuplicates
FROM DimProduct
GROUP BY ProductKey
HAVING COUNT(*) > 1;
```

**Expected Output**: `ProductKey` and `CountOfDuplicates` for any keys appearing more than once in `DimProduct`.

**Business Insight**: Identifies integrity issues in the product dimension, which could cause problems with product identification and reporting.

### Check 5: Duplicate Customer Keys

**Business Explanation**: Similar to products, each customer must have a unique `CustomerKey`. Duplicates can lead to miscounting customers or attributing sales to the wrong entity.

**Hint**: Use `GROUP BY CustomerKey` and `COUNT(*) > 1`.

```sql
SELECT CustomerKey, COUNT(*) AS CountOfDuplicates
FROM DimCustomer
GROUP BY CustomerKey
HAVING COUNT(*) > 1;
```

**Expected Output**: `CustomerKey` and `CountOfDuplicates` for any keys appearing more than once in `DimCustomer`.

**Business Insight**: Ensures accurate customer counts and prevents issues when analyzing customer-specific data, such as total purchases or demographics.

## 🔗 Checks for Referential Integrity (Missing Relationships)

Referential integrity ensures that relationships between tables are maintained. A fact record should always have a corresponding dimension record. If not, it indicates a broken relationship or orphaned fact records.

### Check 6: Fact Records Without Corresponding Customers

**Business Explanation**: Every sale should be associated with an existing customer. Sales without a customer are unidentifiable and impact customer-centric analysis.

**Hint**: Use a `LEFT JOIN` from `FactInternetSales` to `DimCustomer` and check for `DimCustomer.CustomerKey IS NULL`.

```sql
SELECT fis.*
FROM FactInternetSales AS fis
LEFT JOIN DimCustomer AS dc ON fis.CustomerKey = dc.CustomerKey
WHERE dc.CustomerKey IS NULL;
```

**Expected Output**: Any rows from `FactInternetSales` where the `CustomerKey` does not match an existing `CustomerKey` in `DimCustomer`.

**Business Insight**: Highlights sales records that cannot be attributed to a customer, making customer behavior analysis incomplete.

### Check 7: Fact Records Without Corresponding Products

**Business Explanation**: Every sold item in `FactInternetSales` must correspond to an actual product in `DimProduct`. Sales of non-existent products are erroneous.

**Hint**: Use a `LEFT JOIN` from `FactInternetSales` to `DimProduct` and check for `DimProduct.ProductKey IS NULL`.

```sql
SELECT fis.*
FROM FactInternetSales AS fis
LEFT JOIN DimProduct AS dp ON fis.ProductKey = dp.ProductKey
WHERE dp.ProductKey IS NULL;
```

**Expected Output**: Any rows from `FactInternetSales` where the `ProductKey` does not match an existing `ProductKey` in `DimProduct`.

**Business Insight**: Identifies sales of products that are not in the catalog, indicating potential data entry errors or inventory mismatches.

### Check 8: Fact Records Without Corresponding Dates

**Business Explanation**: All sales must have a valid `OrderDateKey` that links to a date in `DimDate`. Missing dates make time-series analysis impossible.

**Hint**: Use a `LEFT JOIN` from `FactInternetSales` to `DimDate` and check for `DimDate.DateKey IS NULL` on the relevant date key (`OrderDateKey`, `DueDateKey`, `ShipDateKey`).

```sql
SELECT fis.*
FROM FactInternetSales AS fis
LEFT JOIN DimDate AS dd ON fis.OrderDateKey = dd.DateKey
WHERE dd.DateKey IS NULL;
```

**Expected Output**: Any rows from `FactInternetSales` where the `OrderDateKey` does not match an existing `DateKey` in `DimDate`.

**Business Insight**: Ensures all sales can be correctly mapped to a calendar date, which is fundamental for any time-based reporting and trend analysis.
