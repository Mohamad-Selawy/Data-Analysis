# Module 1: Database Fundamentals for Analysts

Welcome to the first module of our SQL course! In this section, we'll cover the foundational concepts of databases that are crucial for any aspiring Data Analyst. ## What is a Database?

### Concept Explanation

A **database** is an organized collection of structured information, or data, typically stored electronically in a computer system. It is designed to efficiently store, manage, and retrieve large amounts of data. Think of it like a highly organized digital filing cabinet, where different types of information are stored in specific folders (tables) and within those folders, each document (row) has specific pieces of information (columns).

Databases are fundamental to almost every application and business operation today. As a Data Analyst, you'll spend a significant portion of your time interacting with databases to extract, transform, and analyze data to support business decisions.

### Business Interpretation

For AdventureWorks, a database holds all the critical information about its operations: customer details, product catalogs, sales orders, employee records, vendor information, and much more. Without a robust database, managing a global manufacturing company like AdventureWorks would be impossible. Analysts rely on this centralized data repository to understand business performance, identify trends, and answer critical business questions.

## Relational Database Concepts

### Concept Explanation

A **Relational Database** is the most common type of database used in analytical and business applications. It organizes data into one or more data tables (or 'relations') in which data types may be related to each other. Each table has a unique name and is composed of rows and columns.

The key idea behind relational databases is the establishment of relationships between tables based on common fields. This structure allows for efficient storage and retrieval of complex, interconnected data without duplication.

### Business Interpretation

AdventureWorks, like most enterprise systems, uses a relational database. This means that customer data, product data, and sales data are stored in separate, but related, tables. For example, a sales order record doesn't need to store all of a customer's address details; instead, it just needs a customer ID that links it to the customer's full details in another table. This design ensures data consistency, reduces redundancy, and allows analysts to combine different pieces of information to get a complete picture of the business.

## Tables, Rows, and Columns

### Concept Explanation

In a relational database:

*   **Tables:** These are the primary storage units, often called 'relations'. Each table stores data about a specific entity or concept (e.g., `Customers`, `Products`, `SalesOrders`).
*   **Rows:** Also known as 'records' or 'tuples', each row in a table represents a single, complete entry or instance of the entity. For example, in a `Customers` table, each row would represent one customer.
*   **Columns:** Also known as 'fields' or 'attributes', each column represents a specific piece of information about the entity. For example, in a `Customers` table, columns might include `CustomerID`, `FirstName`, `LastName`, `Email`, `Address`, etc.

## Primary Keys

### Concept Explanation

A **Primary Key (PK)** is a column (or a set of columns) in a table that uniquely identifies each row in that table. It must contain unique values for each row, and it cannot contain NULL values (i.e., it must always have a value).

Think of it like an ID number for every unique item in a list. In AdventureWorks, every product has a unique `ProductKey`, every customer has a unique `CustomerKey`, and every sales order has a unique `SalesOrderNumber` (or similar key for transactions).

**Characteristics of a Primary Key:**
*   **Unique:** No two rows can have the same primary key value.
*   **Non-Null:** A primary key column cannot contain empty or unknown values.
*   **Immutable (usually):** Primary key values should ideally not change over time.

### Business Interpretation

Primary keys are essential for data integrity and for linking related information across different tables. If you want to find specific details about a customer, you'd use their unique `CustomerKey` to retrieve their information quickly and accurately from the `DimCustomer` table. This prevents errors and ensures that each record is distinctly identifiable within the database.

## Foreign Keys

### Concept Explanation

A **Foreign Key (FK)** is a column (or a set of columns) in one table that refers to the Primary Key in another table. The purpose of the foreign key is to establish and enforce a link between two tables, ensuring referential integrity (meaning that relationships between tables remain consistent).

It's how we connect information. For example, if a `SalesOrder` table has a `CustomerKey` column, that `CustomerKey` is a foreign key, and it links back to the `CustomerKey` (primary key) in the `Customers` table.

## Relationships

### Concept Explanation

**Relationships** in a relational database describe how data in one table is connected to data in another table. These connections are established using primary and foreign keys. There are three main types of relationships:

1.  **One-to-One (1:1):** Each record in Table A relates to exactly one record in Table B, and vice-versa. (Less common)
2.  **One-to-Many (1:M):** Each record in Table A can relate to one or more records in Table B, but each record in Table B relates to only one record in Table A. (Most common)
3.  **Many-to-Many (M:M):** Each record in Table A can relate to one or more records in Table B, and vice-versa. (Requires an intermediary 'junction' table).

Understanding relationships is critical for any Data Analyst. ## Fact Tables and Dimension Tables

### Concept Explanation

In the context of data warehousing and analytical databases, tables are often categorized into two main types:

*   **Dimension Tables:** These tables contain descriptive attributes about the business entities. They provide the 'who, what, where, when, why' context for analysis. Dimensions are usually relatively static or change slowly over time. Examples include `DimProduct`, `DimCustomer`, `DimGeography`, `DimDate`.
    *   **Attributes:** The columns in a dimension table are called attributes (e.g., `ProductName`, `CustomerName`, `City`, `CalendarYear`).

*   **Fact Tables:** These tables contain the quantitative measures (metrics) of a business process, along with foreign keys to related dimension tables. Fact tables are typically very large and grow constantly, recording every transaction or event. Examples include `FactInternetSales`, `FactResellerSales`.
    *   **Measures:** The numeric, additive values in a fact table that you want to analyze (e.g., `SalesAmount`, `OrderQuantity`, `Freight`).

### Business Interpretation

This distinction is fundamental for analytical thinking. When AdventureWorks wants to analyze sales, the `FactInternetSales` table holds the actual sales transactions (the 'what happened'). To understand *who* bought *what*, *when*, and *where*, analysts link `FactInternetSales` to dimension tables like `DimCustomer`, `DimProduct`, `DimDate`, and `DimGeography` using foreign keys. This design makes it much easier to perform complex analyses and report on business performance, as it separates descriptive context from measurable events.

## Introduction to Data Warehouse Concepts

### Concept Explanation

A **data warehouse** is a large repository of integrated data collected from various disparate sources, typically used for reporting and data analysis. Unlike operational databases (which are designed for day-to-day transaction processing), data warehouses are optimized for querying and analysis. Key characteristics include:

*   **Subject-Oriented:** Organized around major subjects of the enterprise (e.g., sales, customers, products), rather than business processes.
*   **Integrated:** Data from various sources is combined and harmonized into a consistent format.
*   **Time-Variant:** Data in a data warehouse is tied to specific time periods, allowing for historical analysis.
*   **Non-Volatile:** Once data is entered into the warehouse, it is not updated or deleted; new data is added periodically.

### Business Interpretation

AdventureWorks, being a large enterprise, would undoubtedly utilize a data warehouse. This warehouse would consolidate sales data from different channels (internet, reseller), customer information, product details, and other operational data into a single, clean, and consistent environment. For analysts, this means having a reliable 'single source of truth' for business intelligence and reporting, enabling them to analyze trends over time, compare performance across regions, and make data-driven recommendations without impacting the operational systems.

## Data Warehouse Schemas: Star, Snowflake, and Galaxy

### Concept Explanation
In data warehousing, the way tables are structured is crucial for analysis. Here are three common schema types:

*   **Star Schema:** The simplest model, with a central **fact table** (containing measurements like sales amounts) directly connected to several **dimension tables** (containing descriptive attributes like product details or customer information). It's designed for fast query performance due to fewer joins.

*   **Snowflake Schema:** A more normalized extension of the Star Schema. Dimension tables are further broken down into sub-dimension tables, reducing data redundancy but potentially increasing query complexity due to more joins. For example, a `Product` dimension might link to `ProductSubcategory`, which then links to `ProductCategory`.

*   **Galaxy Schema (Fact Constellation):** This schema involves multiple fact tables that share common dimension tables. It's used for complex data warehouses that need to analyze several related business processes within a single system (e.g., sales and inventory sharing a product dimension).

### AdventureWorks Schema Type

**AdventureWorks primarily utilizes a Snowflake Schema.** You will observe that its dimension tables, such as `DimProduct`, are often normalized into multiple related tables (`DimProduct`, `DimProductSubcategory`, `DimProductCategory`) rather than being a single, flat dimension. This design helps in maintaining data integrity and reducing redundancy, though it means some analytical queries may involve more joins compared to a pure Star Schema.

## SELECT Statement

### Concept Explanation

The `SELECT` statement is used to retrieve data from one or more tables in a database. It's the most frequently used SQL command and forms the basis of almost all data retrieval operations. You specify which columns you want to see, and SQL returns only that data.

*   `SELECT *`: Retrieves all columns from the specified table.
*   `SELECT column1, column2`: Retrieves only the specified columns.

### SQL Syntax

```sql
SELECT column1, column2, ...
FROM table_name;

-- Or to select all columns:
SELECT *
FROM table_name;
```

## WHERE Clause

### Concept Explanation

The `WHERE` clause is used to filter records. It extracts only those records that fulfill a specified condition. This is incredibly powerful for analysts, as it allows you to narrow down vast datasets to only the relevant information needed for a particular business question.

Conditions in the `WHERE` clause can use various operators:

*   **Comparison Operators:** `= (Equal)`, `!= or <> (Not Equal)`, `> (Greater Than)`, `< (Less Than)`, `>= (Greater Than or Equal To)`, `<= (Less Than or Equal To)`.
*   **Logical Operators:** `AND`, `OR`, `NOT`.
*   **Special Operators:** `IN` (for a list of values), `BETWEEN` (for a range), `LIKE` (for pattern matching), `IS NULL` (for checking null values).

### SQL Syntax

```sql
SELECT column1, column2, ...
FROM table_name
WHERE condition;
```

### AdventureWorks Example

Let's find all products from the `DimProduct` table that have a `Color` of 'Red' and a `ListPrice` greater than 1000.

```sql
SELECT
    EnglishProductName,
    Color,
    ListPrice
FROM
    DimProduct
WHERE
    Color = 'Red' AND ListPrice > 1000;
```

### Query Explanation

*   `SELECT EnglishProductName, Color, ListPrice`: Specifies the columns we want to retrieve.
*   `FROM DimProduct`: Indicates that we are querying the `DimProduct` table.
*   `WHERE Color = 'Red' AND ListPrice > 1000`: This is the filtering condition. We are asking the database to return only those rows where the `Color` column has the value 'Red' AND the `ListPrice` column has a value greater than 1000. Both conditions must be true for a row to be included in the result.

### Business Interpretation

This query allows an analyst to quickly identify specific products that meet certain criteria. For AdventureWorks, this could be useful for a marketing campaign targeting high-value red products, or for inventory management to focus on premium red items. Filtering helps in segmenting products for various business strategies without manually sifting through thousands of entries.

## DISTINCT Keyword

### SQL Syntax

## TOP Clause

### Concept Explanation

The `TOP` clause is used to retrieve a specified number of rows or a specified percentage of rows from the beginning of the result set. It's particularly useful when you're interested in only the first few records that meet certain criteria, such as the highest-selling products or the most recent orders.

Note: `TOP` is specific to SQL Server (T-SQL). Other SQL dialects might use `LIMIT` (MySQL, PostgreSQL) or `ROWNUM` (Oracle).

### SQL Syntax

```sql
SELECT TOP number | PERCENT column1, column2, ...
FROM table_name
WHERE condition
ORDER BY column_name [ASC|DESC];
```

### AdventureWorks Example

Let's find the top 5 most expensive products from the `DimProduct` table.

```sql
SELECT TOP 5
    EnglishProductName,
    ListPrice
FROM
    DimProduct
ORDER BY
    ListPrice DESC;
```

### Query Explanation

*   `SELECT TOP 5 EnglishProductName, ListPrice`: This specifies that we want to retrieve the `EnglishProductName` and `ListPrice` for only the top 5 rows.
*   `FROM DimProduct`: We are querying the `DimProduct` table.
*   `ORDER BY ListPrice DESC`: This is crucial! To get the 'most expensive' products, we first need to sort the products by `ListPrice` in descending order. Without `ORDER BY`, `TOP` would return an arbitrary 5 rows.

### Business Interpretation

For AdventureWorks, identifying the top N most expensive products can inform pricing strategies, premium product marketing, or inventory decisions. A Data Analyst might use this to understand the high-end segment of their product offerings, compare them against competitors, or focus sales efforts. This is also useful for quickly sampling data or for reporting on leading indicators.

# Module 3: Aggregations and Business Metrics

In this module, we move from retrieving individual rows to calculating summary statistics. For a Data Analyst, this is where the real value is created—turning raw transaction data into Key Performance Indicators (KPIs).

## Aggregate Functions (COUNT, SUM, AVG, MIN, MAX)

### Concept Explanation
Aggregate functions perform a calculation on a set of values and return a single value.

*   `COUNT()`: Counts the number of rows.
*   `SUM()`: Calculates the total of a numeric column.
*   `AVG()`: Calculates the average value.
*   `MIN()`: Finds the smallest value.
*   `MAX()`: Finds the largest value.

### SQL Syntax
```sql
SELECT
    COUNT(column_name),
    SUM(column_name),
    AVG(column_name)
FROM table_name;
```

### AdventureWorks Example
Let's calculate some basic metrics for our internet sales.

```sql
SELECT
    COUNT(SalesOrderNumber) AS TotalOrders,
    SUM(SalesAmount) AS TotalRevenue,
    AVG(SalesAmount) AS AverageOrderValue,
    MIN(SalesAmount) AS SmallestSale,
    MAX(SalesAmount) AS LargestSale
FROM FactInternetSales;
```

### Query Explanation
*   `COUNT(SalesOrderNumber)`: Counts every row in the sales table to give us the total volume of transactions.
*   `SUM(SalesAmount)`: Adds up every dollar earned from internet sales.
*   `AVG(SalesAmount)`: Calculates the average revenue per transaction line item.

### Business Interpretation
These numbers represent the high-level health of the business. Total Revenue tells you the scale of the company, while Average Order Value (AOV) helps you understand customer spending behavior. If AOV is low, marketing might try to bundle products together to increase ticket sizes.

## GROUP BY Clause

### Concept Explanation
The `GROUP BY` statement groups rows that have the same values into summary rows. It is almost always used with aggregate functions to provide metrics *by category* (e.g., sales by year, sales by product).

### SQL Syntax
```sql
SELECT
    column_name,
    AGGREGATE_FUNCTION(column_name)
FROM table_name
GROUP BY column_name;
```

### AdventureWorks Example
How much revenue did we generate for each unique product key?

```sql
SELECT
    ProductKey,
    SUM(SalesAmount) AS TotalRevenue,
    COUNT(SalesOrderNumber) AS OrderCount
FROM FactInternetSales
GROUP BY ProductKey
ORDER BY TotalRevenue DESC;
```

### Query Explanation
*   `GROUP BY ProductKey`: This tells SQL to collect all rows for Product #1, sum them up, then move to Product #2, and so on.
*   `ORDER BY TotalRevenue DESC`: We sort the results so the highest-earning products appear at the top.

### Business Interpretation
This is a 'Best Selling Products' analysis. By grouping by `ProductKey`, an analyst can identify which items are driving the most revenue versus which items are sold most frequently. This helps the inventory team decide which products to restock more aggressively.

## HAVING Clause

### Concept Explanation
The `HAVING` clause was added to SQL because the `WHERE` keyword cannot be used with aggregate functions. `HAVING` allows you to filter the results *after* the aggregation has occurred.

### SQL Syntax
```sql
SELECT
    column_name,
    SUM(column_name)
FROM table_name
GROUP BY column_name
HAVING SUM(column_name) > value;
```

### AdventureWorks Example
Find products that have generated more than $50,000 in total revenue.

```sql
SELECT
    ProductKey,
    SUM(SalesAmount) AS TotalRevenue
FROM FactInternetSales
GROUP BY ProductKey
HAVING SUM(SalesAmount) > 50000;
```

### Query Explanation
*   `GROUP BY ProductKey`: Groups the sales by product.
*   `HAVING SUM(SalesAmount) > 50000`: This filters the grouped results. If a product's total sum is less than 50k, it is excluded from the final list.

### Business Interpretation
This is a 'High Performer' filter. Analysts use `HAVING` to focus on significant segments of the business—for example, identifying only those regions that are hitting their targets or products that are high-priority earners.

# Module 4: Working with Multiple Tables (Deep Dive into JOINs)

In a relational database, data is spread across multiple tables to reduce redundancy. As a Data Analyst, your primary job is to 'join' these tables back together to answer business questions that span different entities (e.g., 'What is the name of the customer who placed this order?').

## Understanding INNER JOIN

### Concept Explanation
An `INNER JOIN` returns records that have matching values in both tables. If a record in the left table doesn't have a match in the right table, it is excluded from the result.

### SQL Syntax
```sql
SELECT TableA.Column, TableB.Column
FROM TableA
INNER JOIN TableB
ON TableA.CommonKey = TableB.CommonKey;
```

### AdventureWorks Example
Let's join `FactInternetSales` with `DimCustomer` to see the names of customers who actually made purchases.

```sql
SELECT
    S.SalesOrderNumber,
    S.SalesAmount,
    C.FirstName,
    C.LastName
FROM FactInternetSales AS S
INNER JOIN DimCustomer AS C
    ON S.CustomerKey = C.CustomerKey;
```

### Business Interpretation
This query provides a transaction list with human-readable names. Without the join, you only see `CustomerKey` (a number), which means nothing to a business manager. This allows the sales team to follow up with specific individuals.

## LEFT JOIN (Outer Join)

### Concept Explanation
A `LEFT JOIN` returns all records from the left table, and the matched records from the right table. If there is no match, the result is `NULL` on the right side.

### SQL Syntax
```sql
SELECT columns
FROM TableA
LEFT JOIN TableB
ON TableA.Key = TableB.Key;
```

### AdventureWorks Example
Let's find all products and see their sales. Some products might have never been sold.

```sql
SELECT
    P.EnglishProductName,
    S.SalesOrderNumber,
    S.SalesAmount
FROM DimProduct AS P
LEFT JOIN FactInternetSales AS S
    ON P.ProductKey = S.ProductKey
WHERE S.SalesOrderNumber IS NULL;
```

### Business Interpretation
This is a 'Gap Analysis'. By using a `LEFT JOIN` and looking for `NULL` values, we identify products that are in our catalog but have **zero sales**. This is critical for inventory managers to identify 'dead stock' that should perhaps be discontinued or discounted.

## Joining Multiple Tables

### Concept Explanation
You can chain joins together to build a complete picture from many dimensions.

### AdventureWorks Example
Let's calculate total sales by Product Category.

```sql
SELECT
    Cat.EnglishProductCategoryName AS Category,
    SUM(S.SalesAmount) AS TotalRevenue
FROM FactInternetSales AS S
INNER JOIN DimProduct AS P ON S.ProductKey = P.ProductKey
INNER JOIN DimProductSubcategory AS Sub ON P.ProductSubcategoryKey = Sub.ProductSubcategoryKey
INNER JOIN DimProductCategory AS Cat ON Sub.ProductCategoryKey = Cat.ProductCategoryKey
GROUP BY Cat.EnglishProductCategoryName;
```

### Business Interpretation
This query connects 4 tables: Sales -> Product -> Subcategory -> Category. This is the ultimate 'Executive View,' showing which business divisions (Bikes vs. Clothing) are driving the most revenue.

# Module 5: SQL Functions (String, Date, Numeric)

Data is rarely perfect. As an analyst, you'll often need to clean text, calculate the difference between two dates, or round financial figures. SQL provides a robust library of built-in functions to handle these tasks.

## String Functions

### Concept Explanation
String functions allow you to manipulate text data. Common ones include:
*   `UPPER()` / `LOWER()`: Converts text to all caps or all lowercase.
*   `LEN()`: Returns the number of characters in a string.
*   `LEFT()` / `RIGHT()`: Extracts a specific number of characters from the start or end.
*   `CONCAT()` or `+`: Combines two or more strings together.

### AdventureWorks Example
Let's format customer names for a mailing list.

```sql
SELECT
    FirstName,
    LastName,
    UPPER(LastName) AS LastName_Caps,
    CONCAT(FirstName, ' ', LastName) AS FullName,
    LEFT(EmailAddress, 5) AS EmailSnippet
FROM DimCustomer;
```

### Business Interpretation
Consistency is key in customer-facing communication. Analysts use `UPPER` or `LOWER` to standardize user-inputted data, ensuring that 'Smith' and 'smith' are treated the same in reports.

## Date Functions

### Concept Explanation
Since almost all business analysis involves time, date functions are critical:
*   `GETDATE()`: Returns the current system date and time.
*   `YEAR()` / `MONTH()` / `DAY()`: Extracts specific parts of a date.
*   `DATEDIFF()`: Calculates the difference between two dates (e.g., days between order and ship date).
*   `DATEADD()`: Adds a specific interval (e.g., 30 days) to a date.

### AdventureWorks Example
How long does it take, on average, to ship an order?

```sql
SELECT
    SalesOrderNumber,
    OrderDate,
    ShipDate,
    DATEDIFF(day, OrderDate, ShipDate) AS DaysToShip
FROM FactInternetSales;
```

### Business Interpretation
`DATEDIFF` is the foundation of Logistics analysis. By measuring 'DaysToShip', an analyst can identify bottlenecks in the warehouse. If the average shipping time increases, it could lead to poor customer satisfaction.

## Numeric Functions

### Concept Explanation
These functions help with mathematical adjustments:
*   `ROUND()`: Rounds a number to a specified number of decimal places.
*   `ABS()`: Returns the absolute (positive) value of a number.
*   `CEILING()` / `FLOOR()`: Rounds up or down to the nearest integer.

### AdventureWorks Example
Let's clean up our tax and freight calculations.

```sql
SELECT
    SalesOrderNumber,
    TaxAmt,
    ROUND(TaxAmt, 1) AS RoundedTax,
    FLOOR(Freight) AS MinFreightCost
FROM FactInternetSales;
```

### Business Interpretation
Rounding is vital for financial reporting. Executives often want to see high-level numbers without distracting cents. `ROUND` ensures that your SQL results match the totals seen in accounting software.

# Module 6: CASE Statements (Logic and Categorization)

The `CASE` statement is SQL's way of handling `IF-THEN-ELSE` logic. It allows you to create new columns based on conditions, which is essential for data segmentation and creating custom flags for business reporting.

## Basic CASE Statement

### Concept Explanation
The `CASE` statement goes through conditions and returns a value when the first condition is met. If no conditions are met, it returns the value in the `ELSE` clause (or `NULL` if no `ELSE` is provided).

### SQL Syntax
```sql
SELECT
    column_name,
    CASE
        WHEN condition1 THEN result1
        WHEN condition2 THEN result2
        ELSE resultN
    END AS alias_name
FROM table_name;
```

### AdventureWorks Example
Let's categorize customers based on their `YearlyIncome` to help the marketing team with targeting.

```sql
SELECT
    FirstName,
    LastName,
    YearlyIncome,
    CASE
        WHEN YearlyIncome >= 100000 THEN 'High Income'
        WHEN YearlyIncome >= 50000 THEN 'Middle Income'
        ELSE 'Low Income'
    END AS IncomeSegment
FROM DimCustomer;
```

### Business Interpretation
This logic creates a 'Marketing Segment'. Instead of looking at raw income numbers, the business can now simply ask for a count of 'High Income' customers. This makes it much easier to design targeted email campaigns or loyalty programs.

## Using CASE with Aggregations

### Concept Explanation
You can use `CASE` inside an aggregate function like `SUM` or `COUNT` to perform 'pivoting' or conditional counting. This is a very powerful technique for creating complex KPIs in a single query.

### AdventureWorks Example
Let's count how many 'Luxury' vs 'Standard' items were sold. We'll define 'Luxury' as any item with a unit price over $2,000.

```sql
SELECT
    SUM(CASE WHEN UnitPrice > 2000 THEN 1 ELSE 0 END) AS LuxuryItemCount,
    SUM(CASE WHEN UnitPrice <= 2000 THEN 1 ELSE 0 END) AS StandardItemCount
FROM FactInternetSales;
```

### Business Interpretation
This provides a snapshot of the sales mix. If 'LuxuryItemCount' is shrinking compared to 'StandardItemCount', the business might be losing its premium market share, prompting a change in strategy.

# Module 7: Subqueries and CTEs

Sometimes, a single query isn't enough to answer a complex question. You might need to calculate a value first (like an average) and then use that value in a second calculation. This is where Subqueries and Common Table Expressions (CTEs) come in.

## Subqueries

### Concept Explanation
A subquery is a query nested inside another query. It is usually used in the `WHERE` clause to filter data based on the results of a separate search.

### AdventureWorks Example
Find all products that have a list price greater than the average price of all products.

```sql
SELECT
    EnglishProductName,
    ListPrice
FROM DimProduct
WHERE ListPrice > (SELECT AVG(ListPrice) FROM DimProduct);
```

### Business Interpretation
This is a 'Benchmark Analysis'. By using a subquery to find the average, we can dynamically identify premium products without needing to hard-code a specific dollar amount. If prices change tomorrow, the query still works.

## Common Table Expressions (CTEs)

### Concept Explanation
A CTE is a temporary result set that you define at the start of your query using the `WITH` keyword. Think of it as creating a temporary table that exists only for the duration of that specific query. CTEs are preferred over subqueries because they are much easier to read and maintain.

### SQL Syntax
```sql
WITH CTE_Name AS (
    SELECT column1, column2
    FROM TableA
)
SELECT *
FROM CTE_Name;
```

### AdventureWorks Example
Let's find the total sales per customer and then filter for only those customers who have spent more than $5,000.

```sql
WITH CustomerSales AS (
    SELECT
        CustomerKey,
        SUM(SalesAmount) AS TotalSpent
    FROM FactInternetSales
    GROUP BY CustomerKey
)
SELECT
    C.FirstName,
    C.LastName,
    CS.TotalSpent
FROM DimCustomer C
JOIN CustomerSales CS ON C.CustomerKey = CS.CustomerKey
WHERE CS.TotalSpent > 5000
ORDER BY CS.TotalSpent DESC;
```

### Business Interpretation
This structure allows us to separate the *aggregation* (summing sales) from the *descriptive join* (getting customer names). For a Data Analyst, this 'modular' way of writing SQL is standard practice because it prevents errors in complex reporting.

## Temporary Tables (SQL Server specific)

### Concept Explanation

**Temporary tables** are physical tables that are created in the `tempdb` system database and exist only for the duration of the current user session or the stored procedure that created them. They are useful for storing intermediate results of complex queries, breaking down large queries into smaller, more manageable steps, or holding data that needs to be manipulated multiple times within a single batch.

There are two main types:
*   **Local Temporary Tables (`#TableName`):** Visible only to the current session. They are automatically dropped when the session ends.

The `SELECT ... INTO` statement is a common way to quickly create a new table (including temporary tables) and populate it with the results of a query.

### SQL Syntax

```sql
SELECT column1, column2, ...
INTO #TemporaryTableName
FROM SourceTableName
WHERE condition;
```

### AdventureWorks Example

Let's create a temporary table containing information about high-value customers (those with `YearlyIncome` greater than $80,000) and then query this temporary table.

````sql
-- Create the temporary table
SELECT
    CustomerKey,
    FirstName,
    LastName,
    YearlyIncome
INTO #HighValueCustomers
FROM DimCustomer
WHERE YearlyIncome > 80000;

-- Query the temporary table
SELECT *
FROM #HighValueCustomers
ORDER BY YearlyIncome DESC;

-- The temporary table #HighValueCustomers will automatically be dropped when the session ends.

### Business Interpretation

Temporary tables are particularly useful when you need to perform multiple operations on an intermediate result set without affecting the original tables or creating permanent objects in the database. For an analyst, this can streamline complex multi-step analyses, such as creating segmented customer lists for targeted marketing campaigns, or aggregating data in several stages before final reporting. They provide flexibility and can sometimes offer performance benefits by reducing the amount of data processed in subsequent steps.

### Practice Exercises (Module 7)

1.  **Use a Subquery to find all sales in `FactInternetSales` for the product with the highest `UnitPrice`.**
    *   *Business Question:* Who are the customers buying our most expensive single item?

2.  **Create a CTE called `ProductPerformance` that calculates the total `OrderQuantity` for every product. Then, join it with `DimProduct` to show the Product Name and the total units sold.**
    *   *Business Question:* Create a simple inventory movement report.

3.  **Use a CTE to identify the top 10% of customers by income, and then list their names and occupations.**

### Solutions

1.  **Max Price Subquery**
    ```sql
    SELECT SalesOrderNumber, SalesAmount
    FROM FactInternetSales
    WHERE UnitPrice = (SELECT MAX(UnitPrice) FROM FactInternetSales);
    ```

2.  **Product Performance CTE**
    ```sql
    WITH ProductPerformance AS (
        SELECT ProductKey, SUM(OrderQuantity) AS TotalUnits
        FROM FactInternetSales
        GROUP BY ProductKey
    )
    SELECT P.EnglishProductName, PP.TotalUnits
    FROM DimProduct P
    JOIN ProductPerformance PP ON P.ProductKey = PP.ProductKey;
    ```

3.  **Top Income CTE**
    ```sql
    WITH IncomeRank AS (
        SELECT FirstName, LastName, EnglishOccupation, YearlyIncome
        FROM DimCustomer
    )
    SELECT TOP 10 PERCENT *
    FROM IncomeRank
    ORDER BY YearlyIncome DESC;
    ```

# Module 8: Window Functions (Ranking, Running Totals)

Window functions (also known as Analytical Functions) perform a calculation across a set of table rows that are somehow related to the current row. Unlike regular aggregate functions, window functions do not cause rows to become grouped into a single output row — the rows retain their separate identities.

## The OVER() Clause and PARTITION BY

### Concept Explanation
The `OVER()` clause defines a window or user-specified set of rows within a query result set.
*   `PARTITION BY`: Divides the result set into partitions (groups). The window function is applied to each partition separately.
*   `ORDER BY`: Defines the logical order of the rows within each partition.

### SQL Syntax
```sql
SELECT
    column_name,
    AGGREGATE_FUNCTION(column_name) OVER (PARTITION BY column_name ORDER BY column_name) AS Alias
FROM table_name;
```

### AdventureWorks Example
Let's calculate a running total of sales for each customer.

```sql
SELECT
    CustomerKey,
    OrderDate,
    SalesAmount,
    SUM(SalesAmount) OVER (PARTITION BY CustomerKey ORDER BY OrderDate) AS RunningTotal
FROM FactInternetSales
ORDER BY CustomerKey, OrderDate;
```

### Business Interpretation
Running totals are essential for 'Velocity Analysis'. By seeing how a customer's spending accumulates over time, an analyst can identify when a customer is becoming more (or less) engaged with the brand, allowing for proactive outreach.

### AdventureWorks Example: Average Sales Per Customer Using Temporary Table

Let's calculate the average sales amount for each customer across all their orders. We'll first create a temporary table to aggregate sales per order and then apply the window function.

````sql
-- Create a temporary table to store total sales per order for each customer
SELECT CustomerKey, OrderDate, SalesOrderNumber, SUM(SalesAmount) AS Total_Sales
INTO #Customer_Sales
FROM FactInternetSales
GROUP BY CustomerKey, OrderDate, SalesOrderNumber;

````sql
-- Now, use the temporary table to calculate the average sales per customer
SELECT
    CustomerKey,
    OrderDate,
    Total_Sales,
    AVG(Total_Sales) OVER (
        PARTITION BY CustomerKey
    ) AS AverageSalesPerCustomer
FROM #Customer_Sales
ORDER BY CustomerKey, OrderDate;

### Business Interpretation

This example is useful for 'Customer Behavior Analysis'. By calculating the average sales per customer across their transactions, an analyst can benchmark individual order values against a customer's typical spending. This can help identify larger-than-average purchases that might be influenced by specific promotions, or smaller purchases that could indicate a change in buying habits, allowing for targeted interventions or personalized offers. The use of a temporary table also demonstrates how complex multi-step analyses can be broken down for clarity and potentially performance.

## Ranking Functions (RANK and DENSE_RANK)

### Concept Explanation
Ranking functions allow you to assign a rank to each row within a partition.
*   `RANK()`: Assigns a rank. If there is a tie, it leaves a gap in the numbering sequence.
*   `DENSE_RANK()`: Assigns a rank. If there is a tie, it does NOT leave a gap.

### AdventureWorks Example
Who are our top 3 most expensive products within each category?

```sql
WITH RankedProducts AS (
    SELECT
        P.EnglishProductName,
        C.EnglishProductCategoryName,
        P.ListPrice,
        DENSE_RANK() OVER (PARTITION BY C.EnglishProductCategoryName ORDER BY P.ListPrice DESC) AS PriceRank
    FROM DimProduct P
    JOIN DimProductSubcategory S ON P.ProductSubcategoryKey = S.ProductSubcategoryKey
    JOIN DimProductCategory C ON S.ProductCategoryKey = C.ProductCategoryKey
)
SELECT *
FROM RankedProducts
WHERE PriceRank <= 3;
```

### Business Interpretation
Ranking is used for 'Competitive Benchmarking'. By ranking products within their categories, a product manager can see which items are their 'premium' offerings and ensure they are marketed accordingly.

### Practice Exercises (Module 8)

1.  **Use `ROW_NUMBER()` to assign a unique number to every sale in `FactInternetSales`, ordered by `OrderDate` descending.**
    *   *Business Question:* Get a list of our most recent transactions numbered 1 to N.

2.  **Rank sales territories (`DimSalesTerritory`) by their total `SalesAmount` using `RANK()`.**

### Solutions

1.  **Row Numbering**
    ```sql
    SELECT
        SalesOrderNumber,
        OrderDate,
        ROW_NUMBER() OVER (ORDER BY OrderDate DESC) AS TransactionID
    FROM FactInternetSales;
    ```

2.  **Territory Ranking**
    ```sql
    WITH TerritorySales AS (
        SELECT
            T.SalesTerritoryRegion,
            SUM(S.SalesAmount) AS TotalRevenue
        FROM FactInternetSales S
        JOIN DimSalesTerritory T ON S.SalesTerritoryKey = T.SalesTerritoryKey
        GROUP BY T.SalesTerritoryRegion
    )
    SELECT
        SalesTerritoryRegion,
        TotalRevenue,
        RANK() OVER (ORDER BY TotalRevenue DESC) AS TerritoryRank
    FROM TerritorySales;
    ```

# Module 9: Data Quality Checks

Before an analyst can build a dashboard or present a finding, they must ensure the data is 'clean'. Data quality checks are the SQL queries you run to find errors, missing information, or inconsistencies that could lead to incorrect business decisions.

## Identifying NULL Values

### Concept Explanation
NULL represents a missing or unknown value. While some NULLs are acceptable (like a customer not having a middle name), others can break calculations (like a missing price).

### AdventureWorks Example
Let's check if any products in our catalog are missing a `ModelName` or `ListPrice`.

```sql
SELECT
    COUNT(*) AS TotalProducts,
    SUM(CASE WHEN ModelName IS NULL THEN 1 ELSE 0 END) AS MissingModelNames,
    SUM(CASE WHEN ListPrice IS NULL THEN 1 ELSE 0 END) AS MissingPrices
FROM DimProduct;
```

### Business Interpretation
If a product has a NULL price, it cannot be sold. Identifying these 'data gaps' allows the analyst to notify the product team to update the database before the next sales cycle begins.

## Checking for Duplicates

### Concept Explanation
Duplicate records can lead to 'double counting' revenue or overestimating customer counts. We use `GROUP BY` and `HAVING COUNT(*) > 1` to find them.

### AdventureWorks Example
Are there any customers with the same email address in our system?

```sql
SELECT
    EmailAddress,
    COUNT(*) AS OccurrenceCount
FROM DimCustomer
GROUP BY EmailAddress
HAVING COUNT(*) > 1;
```

### Business Interpretation
Duplicate emails usually indicate that a customer registered twice. This skews 'Total Customer' counts and can lead to a customer receiving the same marketing email twice, which is a poor user experience.

## Relationship Integrity Checks

### Concept Explanation
When joining tables, we want to ensure we aren't losing data. An 'Orphaned Record' check looks for foreign keys that don't have a matching primary key.

### AdventureWorks Example
Are there any sales in `FactInternetSales` that point to a product that doesn't exist in `DimProduct`?

```sql
SELECT DISTINCT S.ProductKey
FROM FactInternetSales S
LEFT JOIN DimProduct P ON S.ProductKey = P.ProductKey
WHERE P.ProductKey IS NULL;
```

### Business Interpretation
If this query returns results, it means our database has 'broken' links. We are recording sales for ghost products. This is a major data integrity issue that requires immediate attention from the Database Administrator (DBA).

### Practice Exercises (Module 9)

1.  **Check `DimCustomer` for any records where both `FirstName` and `LastName` are NULL.**
2.  **Identify if any `SalesOrderNumber` appears more than once in `FactInternetSales` for the same `SalesOrderLineNumber`.**
3.  **Check for 'Orphaned' customers: Are there any orders in `FactInternetSales` where the `CustomerKey` does not exist in `DimCustomer`?**

### Solutions

1.  **Missing Names**
    ```sql
    SELECT *
    FROM DimCustomer
    WHERE FirstName IS NULL AND LastName IS NULL;
    ```

2.  **Duplicate Transaction Lines**
    ```sql
    SELECT SalesOrderNumber, SalesOrderLineNumber, COUNT(*)
    FROM FactInternetSales
    GROUP BY SalesOrderNumber, SalesOrderLineNumber
    HAVING COUNT(*) > 1;
    ```

3.  **Orphaned Customers**
    ```sql
    SELECT DISTINCT S.CustomerKey
    FROM FactInternetSales S
    LEFT JOIN DimCustomer C ON S.CustomerKey = C.CustomerKey
    WHERE C.CustomerKey IS NULL;
    ```

# Module 10: Views (Simplifying Complex Queries)

## Concept Explanation

A **View** is a virtual table based on the result-set of an SQL query. A view contains rows and columns, just like a real table. The fields in a view are fields from one or more real tables in the database.

Views do not store data themselves; they merely store the query that defines them. When you query a view, the underlying SQL query is executed, and the results are presented as if they came from a single table.

### Why use Views?
*   **Simplify complex queries:** By encapsulating complex `JOIN`s and `WHERE` clauses into a single view.
*   **Enhance security:** Users can be granted access to specific data through a view without giving them direct access to the underlying tables.
*   **Provide data abstraction:** Presenting data in a way that is logical for a user, independent of how the data is physically stored.

### SQL Syntax

```sql
CREATE VIEW ViewName AS
SELECT column1, column2, ...
FROM table_name
WHERE condition;

-- To query a view:
SELECT * FROM ViewName;

-- To modify a view:
ALTER VIEW ViewName AS
SELECT new_column1, new_column2, ...
FROM new_table_name
WHERE new_condition;

-- To drop a view:
DROP VIEW ViewName;
```

### AdventureWorks Example

Let's create a view that shows the `EnglishProductName`, `SalesOrderNumber`, and `SalesAmount` for all sales, but only for products in the 'Bikes' category. This view will simplify querying sales data for bikes without repeatedly joining tables and applying filters.

````sql
-- Create the view
CREATE VIEW BikeSales AS
SELECT
    P.EnglishProductName,
    S.SalesOrderNumber,
    S.SalesAmount
FROM FactInternetSales AS S
INNER JOIN DimProduct AS P ON S.ProductKey = P.ProductKey
INNER JOIN DimProductSubcategory AS Sub ON P.ProductSubcategoryKey = Sub.ProductSubcategoryKey
INNER JOIN DimProductCategory AS Cat ON Sub.ProductCategoryKey = Cat.ProductCategoryKey
WHERE Cat.EnglishProductCategoryName = 'Bikes';

-- Now, query the view as if it were a table
SELECT *
FROM BikeSales
WHERE SalesAmount > 1000
ORDER BY SalesAmount DESC;
````

### Business Interpretation

Views are incredibly powerful for creating simplified, user-friendly data sets for various stakeholders. For AdventureWorks, a marketing manager interested only in bike sales can now query `BikeSales` without needing to understand the complex joins between `FactInternetSales`, `DimProduct`, `DimProductSubcategory`, and `DimProductCategory`. This reduces the chances of errors, makes ad-hoc reporting faster, and allows analysts to focus on interpreting the data rather than writing repetitive complex queries. It ensures that different departments are always looking at a consistent, pre-defined subset of data relevant to their needs.
