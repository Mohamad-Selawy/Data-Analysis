# NumPy

Welcome to this practical hands-on notebook! In this session, you will practice using **NumPy**, one of Python's most powerful libraries for scientific computing and data analysis.

### Objective
Using a dataset of supermarket transactions, you will learn and apply fundamental NumPy operations such as:
* Loading datasets with NumPy
* Array slicing and 2D indexing
* Aggregation operations (`sum`, `mean`, `max`)
* Boolean indexing (filtering data based on conditions)

Let's get started!

## 1. Import Libraries and Load Data

First, we need to import NumPy and load our dataset.

We will be using the dataset `supermarket_transactions_numeric.csv`.

Now, we will read the uploaded file using NumPy's `np.loadtxt()`. Since our CSV file contains numerical data but has a header row, we will set `skiprows=1` and specify a comma `,` as our delimiter.

```python
import numpy as np
np.set_printoptions(suppress=True)
np.set_printoptions(linewidth=1000)

# Load dataset into a NumPy array
data = np.loadtxt('supermarket_transactions_numeric.csv', delimiter=',', skiprows=1)
# Print the shape of our dataset
print(f"Data Shape: {data.shape}")
print("First 5 rows of data:")
print(data[:5])
```

### Column Index Mapping
To reference specific columns in our 2D array, we must use their column indices. Here is your index cheat-sheet for this notebook:

* **Index `0`**: `Transaction_ID`
* **Index `1`**: `Customer_ID`
* **Index `2`**: `Product_ID`
* **Index `3`**: `Store_ID`
* **Index `4`**: `Day`
* **Index `5`**: `Hour`
* **Index `6`**: `Quantity`
* **Index `7`**: `Unit_Price`
* **Index `8`**: `Discount_Rate`
* **Index `9`**: `Payment_Method_Code` (1 = Cash, 2 = Card, 3 = Mobile)
* **Index `10`**: `Gross_Total`
* **Index `11`**: `Net_Total`

## 2. Practical Exercises

---
### Exercise 1: Explore the Dataset

**Goal**:
1. Display the first 10 transactions from the dataset.
2. Print the exact number of transactions (rows) and columns using `.shape` property dynamically.

```python
# Displaying first 10 transactions
first_10 = data[:10]
print("First 10 transactions:")
print(first_10)

# Get shape details
num_rows, num_cols = data.shape
print(f"\nNumber of transactions: {num_rows}")
print(f"Number of columns: {num_cols}")
```

---
### Exercise 2: Total Revenue

**Goal**:
Calculate the total overall revenue generated by the supermarket using the `Net_Total` column (Index `11`).

```python
# Extract the Net_Total column (index 11)
net_totals = data[:, 11]

# Sum up the net totals
total_revenue = net_totals.sum()
print(f"Total Revenue: ${total_revenue:,.2f}")
```

---
### Exercise 3: Average Transaction Value

**Goal**:
Calculate the average transaction value of all recorded purchases using the `Net_Total` column.

```python
# Extract the Net_Total column and find the mean
average_transaction = data[:, 11].mean()
print(f"Average Transaction Value: ${average_transaction:.2f}")
```

---
### Exercise 4: Highest Transaction

**Goal**:
1. Find the maximum value present in the `Net_Total` column.
2. Find the corresponding `Transaction_ID` (Index `0`) for this purchase. (Hint: Use `np.argmax()` to locate the index of the maximum element).

```python
net_totals = data[:, 11]

# Find maximum value
max_value = net_totals.max()

print(f"Highest Transaction Net Total: ${max_value:.2f}")
print("Transaction for this order:")
print(data[data[:,11] == max_value])
```

---
### Exercise 5: Quantity Sold

**Goal**:
1. Calculate the total quantity of products sold using the `Quantity` column (Index `6`).
2. Calculate the average quantity purchased per transaction.

```python
quantities = data[:, 6]

total_quantity = quantities.sum()
average_quantity = quantities.mean()

print(f"Total Quantity Sold: {int(total_quantity)} items")
print(f"Average Items per Transaction: {average_quantity:.2f}")
```

---
### Exercise 6: Filter Expensive Transactions

**Goal**:
1. Use Boolean indexing to filter all transactions where the `Net_Total` is greater than 100.
2. Print the total count of these expensive transactions.

```python
# Create boolean mask
expensive_mask = data[:, 11] > 100

# Filter the dataset using the mask
expensive_transactions = data[expensive_mask]

# Count matching transactions
count_expensive = expensive_transactions.shape[0]

print(f"Number of transactions greater than $100: {count_expensive}")
print("\nSample of these expensive transactions:")
print(expensive_transactions[:3])
```
