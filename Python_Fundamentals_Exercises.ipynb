### Exercise 2.2: `if-elif-else` for Grade Classification

Write a Python program that takes a `score` (integer from 0-100) as input and prints the corresponding letter grade based on these rules:

- `A`: 90-100
- `B`: 80-89
- `C`: 70-79
- `D`: 60-69
- `F`: Below 60

Test with `score = 95`, `score = 82`, `score = 70`, `score = 55`.

```python
# Your solution here
```

### Solution

```python
def classify_grade(score):
    if score >= 90:
        print(f"Score {score}: Grade A")
    elif score >= 80:
        print(f"Score {score}: Grade B")
    elif score >= 70:
        print(f"Score {score}: Grade C")
    elif score >= 60:
        print(f"Score {score}: Grade D")
    else:
        print(f"Score {score}: Grade F")

classify_grade(95)
classify_grade(82)
classify_grade(70)
classify_grade(55)

# Explanation: This demonstrates the use of `if-elif-else` to handle multiple conditions sequentially. The order of conditions is important here.
```

### Exercise 2.3: Checking Multiple Conditions

Write a program that determines if a given year is a "Leap Year" or a "Common Year". A leap year occurs every 4 years, except for years divisible by 100 but not by 400.

- A year is a leap year if it is divisible by 4.
- However, if the year is also divisible by 100, it is NOT a leap year.
- Unless the year is also divisible by 400, then it IS a leap year.

Test with `year = 2000`, `year = 1900`, `year = 2024`, `year = 2023`.

```python
# Your solution here
```

### Solution

```python
def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(f"{year} is a Leap Year.")
    else:
        print(f"{year} is a Common Year.")

is_leap_year(2000) # Expected: Leap Year
is_leap_year(1900) # Expected: Common Year
is_leap_year(2024) # Expected: Leap Year
is_leap_year(2023) # Expected: Common Year

# Explanation: This exercise uses nested conditions (implied by the `and`/`or` logic) within an `if-else` statement to apply multiple criteria for classification.
```

### Exercise 3.2: Counting Vowels in a String with `for` loop and `continue`

Given a string: `text = "Python Programming for Data Analysis"`

Use a `for` loop to count the number of vowels (a, e, i, o, u, case-insensitive) in the string. Use the `continue` statement to skip non-alphabetic characters or consonants. Print the total count of vowels.

```python
# Your solution here
```

### Solution

```python
text = "Python Programming for Data Analysis"
vowels = "aeiou"
vowel_count = 0

for char in text:
    char_lower = char.lower()
    if char_lower not in vowels:
        continue  # Skip if not a vowel
    vowel_count += 1

print(f"The number of vowels is: {vowel_count}")

# Explanation: This exercise illustrates using a `for` loop to iterate over characters in a string, and the `continue` statement to efficiently skip elements that don't meet a certain condition, thereby avoiding unnecessary processing within the loop.
```

### Exercise 3.3: Countdown with `while` loop and `break`

Write a program that performs a countdown from `10` down to `1`. Use a `while` loop. If the current number in the countdown is `5`, print a special message "Halfway there!" and then immediately stop the countdown using `break`.

Print each number in the countdown.

```python
# Your solution here
```

### Solution

```python
count = 10

while count >= 1:
    print(count)
    if count == 5:
        print("Halfway there!")
        break  # Stop the loop
    count -= 1

print("Countdown finished (or stopped early)!")

# Explanation: This exercise demonstrates a `while` loop for repeated execution based on a condition, and the `break` statement to exit the loop prematurely when a specific condition is met.
```

## 4. Functions

### Exercise 4.1: Function to Calculate Average

Write a Python function named `calculate_average` that takes a list of numbers as an argument and returns their average. If the list is empty, it should return `0`.

Test with `numbers1 = [10, 20, 30]` and `numbers2 = []`.

```python
# Your solution here
```

### Solution

```python
def calculate_average(numbers_list):
    if not numbers_list:
        return 0
    return sum(numbers_list) / len(numbers_list)

numbers1 = [10, 20, 30]
numbers2 = []

print(f"Average of {numbers1}: {calculate_average(numbers1)}")
print(f"Average of {numbers2}: {calculate_average(numbers2)}")

# Explanation: This function demonstrates defining a function that takes an argument, performs calculations, and returns a value. It also includes basic error handling for an empty list.
```

### Exercise 4.3: Function to Calculate Factorial

Write a recursive function named `factorial` that calculates the factorial of a non-negative integer `n`. The factorial of a number is the product of all integers from 1 to that number (e.g., `5! = 5 * 4 * 3 * 2 * 1 = 120`). The factorial of 0 is 1.

Test with `n = 0`, `n = 1`, `n = 5`.

```python
# Your solution here
```

### Solution

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

print(f"Factorial of 0: {factorial(0)}")
print(f"Factorial of 1: {factorial(1)}")
print(f"Factorial of 5: {factorial(5)}")

# Explanation: This exercise introduces a recursive function, which calls itself to solve smaller sub-problems. It's an important concept for understanding how functions can be structured for certain types of problems.
```
