# 3. Creating DAX Measures

## Learning Objective

This section focuses on writing Data Analysis Expressions (DAX) to create robust and flexible measures. You will develop key performance indicators (KPIs) and time intelligence calculations essential for business reporting, ensuring accuracy and consistency across your dashboard.

## Tasks

### Core Sales Measures

-   **Total Sales:** Sum of `SalesAmount` from `FactInternetSales`.
    ```dax
    Total Sales = SUM(FactInternetSales[SalesAmount])
    ```
-   **Total Cost:** Sum of `TotalProductCost` from `FactInternetSales`.
    ```dax
    Total Cost = SUM(FactInternetSales[TotalProductCost])
    ```
-   **Total Profit:** `Total Sales - Total Cost`.
    ```dax
    Total Profit = [Total Sales] - [Total Cost]
    ```
-   **Profit Margin %:** `Total Profit / Total Sales`.
    ```dax
    Profit Margin % = DIVIDE([Total Profit], [Total Sales])
    ```
-   **Total Orders:** Count of distinct `SalesOrderNumber`.
    ```dax
    Total Orders = DISTINCTCOUNT(FactInternetSales[SalesOrderNumber])
    ```

### Customer and Order Measures

-   **Total Customers:** Count of distinct `CustomerKey` from `DimCustomer` (or `FactInternetSales`).
    ```dax
    Total Customers = DISTINCTCOUNT(FactInternetSales[CustomerKey])
    ```
-   **Average Order Value:** `Total Sales / Total Orders`.
    ```dax
    Average Order Value = DIVIDE([Total Sales], [Total Orders])
    ```

### Time Intelligence Measures

-   **Sales YTD (Year-to-Date):** Total Sales from the beginning of the year up to the current filter context.
    ```dax
    Sales YTD = TOTALYTD([Total Sales], 'DimDate'[FullDateAlternateKey])
    ```
-   **Sales Last Year:** Total Sales for the same period in the previous year.
    ```dax
    Sales Last Year = CALCULATE([Total Sales], SAMEPERIODLASTYEAR('DimDate'[FullDateAlternateKey]))
    ```
-   **YoY Growth %:** Percentage change of Sales YTD compared to Sales Last Year.
    ```dax
    YoY Growth % = DIVIDE([Sales YTD] - [Sales Last Year], [Sales Last Year])
    ```

### Dynamic Measures

-   **Top Product:** A measure to identify the highest selling product based on `Total Sales` in the current filter context.
    ```dax
    Top Product =
    VAR TopSellingProductKey =
        CALCULATE(
            VALUES(DimProduct[EnglishProductName]),
            TOPN(1, ALLSELECTED(DimProduct[EnglishProductName]), [Total Sales])
        )
    RETURN
        TopSellingProductKey
    ```
-   **Dynamic Report Title:** A text measure that changes based on selected filters (e.g., displaying the selected year).
    ```dax
    Dynamic Report Title =
    VAR SelectedYear = SELECTEDVALUE('DimDate'[CalendarYear])
    RETURN
        IF(ISBLANK(SelectedYear),
            "AdventureWorks Sales Overview",
            "AdventureWorks Sales - Year " & SelectedYear
        )
    ```

## Explanation

DAX measures are the backbone of analytical reports in Power BI, enabling complex calculations beyond simple sums and averages. By defining these measures once, they can be reused across multiple visuals and pages, ensuring consistency and accuracy. Time intelligence functions are particularly powerful for comparing performance over different periods, which is vital for strategic decision-making.

## Checkpoint
