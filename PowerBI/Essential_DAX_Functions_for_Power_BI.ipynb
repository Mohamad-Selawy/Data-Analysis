# Essential DAX Functions for Power BI

A Study Guide: Data Analysis Expressions (DAX) for the Microsoft Power BI

# Introduction
DAX (Data Analysis Expressions) is the formula language used in Power BI, Power Pivot, and Analysis Services to build measures, calculated columns, and calculated tables. It is used to analyze data and create dynamic calculations that respond to filters and selections in a report.

This guide summarizes the most important DAX functions and concepts, categorized for easier understanding.

# 1. Aggregation Functions

These functions perform calculations over a set of rows and return a single scalar value.

## SUM
**Purpose**: Calculates the sum of a column.
**Syntax**:
````dax
SUM(<column>)
````
**Example**: Calculate the total sales amount.
````dax
Total Sales = SUM(Sales[SalesAmount])
````

## AVERAGE
**Purpose**: Calculates the average (arithmetic mean) of a column.
**Syntax**:
````dax
AVERAGE(<column>)
````
**Example**: Calculate the average sales amount.
````dax
Average Sales = AVERAGE(Sales[SalesAmount])
````

## MIN
**Purpose**: Returns the smallest value in a column.
**Syntax**:
````dax
MIN(<column>)
````
**Example**: Find the minimum order quantity.
````dax
Min Order Quantity = MIN(Sales[OrderQuantity])
````

## MAX
**Purpose**: Returns the largest value in a column.
**Syntax**:
````dax
MAX(<column>)
````
**Example**: Find the maximum order quantity.
````dax
Max Order Quantity = MAX(Sales[OrderQuantity])
````

## COUNT
**Purpose**: Counts the number of non-blank values in a column.
**Syntax**:
````dax
COUNT(<column>)
````
**Example**: Count the number of sales records.
````dax
Number of Sales Records = COUNT(Sales[SalesAmount])
````

## COUNTROWS
**Purpose**: Counts the number of rows in a table.
**Syntax**:
````dax
COUNTROWS(<table>)
````
**Example**: Count the number of transactions in the Sales table.
````dax
Number of Transactions = COUNTROWS(Sales)
````

## DISTINCTCOUNT
**Purpose**: Counts the number of distinct (unique) values in a column.
**Syntax**:
````dax
DISTINCTCOUNT(<column>)
````
**Example**: Count the number of unique products sold.
````dax
Unique Products Sold = DISTINCTCOUNT(Sales[ProductID])
````

# 2. Iterating Functions (X-Functions)

These functions iterate over a table and perform a calculation for each row, then aggregate the results. They are more flexible than standard aggregation functions as they allow row-context calculations.

## SUMX
**Purpose**: Evaluates an expression for each row of a table and sums the resulting values.
**Syntax**:
````dax
SUMX(<table>, <expression>)
````
**Example**: Calculate total sales considering discounts.
````dax
Total Sales with Discount = SUMX(Sales, Sales[Price] * Sales[Quantity] * (1 - Sales[Discount]))
````

## AVERAGEX
**Purpose**: Evaluates an expression for each row of a table and calculates the average of the resulting values.
**Syntax**:
````dax
AVERAGEX(<table>, <expression>)
````
**Example**: Calculate the average profit per product.
````dax
Average Profit per Product = AVERAGEX(Products, Products[SalesPrice] - Products[CostPrice])
````

## COUNTX
**Purpose**: Counts the number of non-blank values in an expression evaluated for each row of a table.
**Syntax**:
````dax
COUNTX(<table>, <expression>)
````
**Example**: Count the number of orders with a quantity greater than 10.
````dax
Orders with Quantity > 10 = COUNTX(Sales, IF(Sales[Quantity] > 10, 1))
````

# 3. Logical Functions

These functions allow you to create conditional logic within your DAX expressions.

## IF
**Purpose**: Checks a condition and returns one value when true, and another when false.
**Syntax**:
````dax
IF(<condition>, <value_if_true>, <value_if_false>)
````
**Example**: Categorize sales as 'High' or 'Low'.
````dax
Sales Category = IF(Sales[SalesAmount] > 1000, "High", "Low")
````

## AND
**Purpose**: Checks whether two or more conditions are all true at the same time.
**Syntax**:
````dax
AND(<condition1>, <condition2>...)
````
**Example**: Identify premium customers with high sales.
````dax
Premium High Sales = IF(AND(Customers[IsPremium] = TRUE(), Sales[SalesAmount] > 5000), "Yes", "No")
````

## OR
**Purpose**: Checks whether at least one of several conditions is true.
**Syntax**:
````dax
OR(<condition1>, <condition2>...)
````
**Example**: Identify sales from either Store A or Store B.
````dax
Sales from Specific Stores = IF(OR(Sales[StoreID] = "A", Sales[StoreID] = "B"), "Yes", "No")
````

## NOT
**Purpose**: Reverses the result of a given condition.
**Syntax**:
````dax
NOT(<condition>)
````
**Example**: Identify non-discounted sales.
````dax
Non-Discounted Sales = IF(NOT(Sales[Discount] > 0), "Yes", "No")
````

# 4. Text Functions

These functions manipulate text strings.

## CONCATENATE
**Purpose**: Joins text from different cells/columns together.
**Syntax**:
````dax
CONCATENATE(<text1>, <text2>)
````
**Example**: Create a full product name.
````dax
Full ProductName = CONCATENATE(Products[Brand], " ", Products[Name])
````

## LEFT
**Purpose**: Extracts a specified number of characters from the left of a text string.
**Syntax**:
````dax
LEFT(<text>, <num_chars>)
````
**Example**: Extract product category code.
````dax
Category Code = LEFT(Products[ProductCode], 3)
````

## RIGHT
**Purpose**: Extracts a specified number of characters from the right of a text string.
**Syntax**:
````dax
RIGHT(<text>, <num_chars>)
````
**Example**: Extract the last 4 digits of a product ID.
````dax
Last 4 ProductID = RIGHT(Products[ProductID], 4)
````

## MID
**Purpose**: Extracts a substring of a specified length from a text string, starting at a given position.
**Syntax**:
````dax
MID(<text>, <start_pos>, <num_chars>)
````
**Example**: Extract the middle part of an order ID.
````dax
Order Middle Part = MID(Orders[OrderID], 4, 3)
````

## LEN
**Purpose**: Returns the number of characters in a text string.
**Syntax**:
````dax
LEN(<text>)
````
**Example**: Find the length of product descriptions.
````dax
Description Length = LEN(Products[Description])
````

## SEARCH
**Purpose**: Returns the starting position of one text string within another text string (case-insensitive).
**Syntax**:
````dax
SEARCH(<find_text>, <within_text>, [<start_num>], [<not_found_value>])
````
**Example**: Find if 'laptop' is in the product name.
````dax
Is black = IF(SEARCH("BLACK",DimProduct[EnglishProductName],1,0) > 0, "BLACK","NOT BLACK")
````

## REPLACE
**Purpose**: Replaces part of a text string with a different text string.
**Syntax**:
````dax
REPLACE(<old_text>, <start_num>, <num_chars>, <new_text>)
````
**Example**: Replace a product name prefix.
````dax
Updated Product Name = REPLACE(DimProduct[EnglishProductName], 1, 2, "NEW ")
````

## SUBSTITUTE
**Purpose**: Replaces all occurrences of old text with new text in a string (case-sensitive).
**Syntax**:
````dax
SUBSTITUTE(<text>, <old_text>, <new_text>, [<instance_num>])
````
**Example**: Replace "Inc." with "Incorporated" in company names.
````dax
Adjusted Product Name = SUBSTITUTE(DimProduct[EnglishProductName], "Flat", "Plain")
````

# 5. Date and Time Functions

These functions work with date and time values. DAX stores dates as datetime values.

## CALENDAR
**Purpose**: Generates a table with a single column of dates between a start and end date.
**Syntax**:
````dax
CALENDAR(<start_date>, <end_date>)
````
**Example**: Create a date table for a specific period.
````dax
Date Table = CALENDAR(DATE(2023, 1, 1), DATE(2023, 12, 31))
````

## DATE
**Purpose**: Returns the datetime representation of a date.
**Syntax**:
````dax
DATE(<year>, <month>, <day>)
````
**Example**: Create a specific date.
````dax
Specific Date = DATE(2024, 7, 15)
````

## TODAY
**Purpose**: Returns the current date.
**Syntax**:
````dax
TODAY()
````
**Example**: Get today's date.
````dax
Today's Date = TODAY()
````

## NOW
**Purpose**: Returns the current date and time.
**Syntax**:
````dax
NOW()
````
**Example**: Get current date and time.
````dax
Current Date & Time = NOW()
````

## YEAR
**Purpose**: Returns the year as a four-digit integer.
**Syntax**:
````dax
YEAR(<date>)
````
**Example**: Extract the year from an order date.
````dax
Order Year = YEAR(Orders[OrderDate])
````

## MONTH

````dax
MONTH(<date>)
````
**Example**: Extract the month from an order date.

## DAY

**Syntax**:
````dax
DAY(<date>)
````
**Example**: Extract the day from an order date.

## HOUR
**Purpose**: Returns the hour as a number from 0 (12:00 AM) to 23 (11:00 PM).
**Syntax**:
````dax
HOUR(<datetime>)
````

**Example**: Extract the minute from a transaction timestamp.

## SECOND
**Purpose**: Returns the second as a number from 0 to 59.
**Syntax**:
````dax
SECOND(<datetime>)
````
**Example**: Extract the second from a transaction timestamp.

# 6. Filter Functions

These functions allow you to modify the context in which data is evaluated. They are fundamental to creating powerful and dynamic calculations in DAX.

## CALCULATE
**Purpose**: Changes the context in which the data is evaluated, allowing you to modify filters. This is one of the most powerful functions in DAX.
**Syntax**:
````dax
CALCULATE(<expression>, [<filter1>], [<filter2>]...)
````
**Example**: Calculate total sales for a specific product category.
````dax
Sales for Bikes = CALCULATE(SUM(FactInternetSales[SalesAmount]), DimProductCategory[EnglishProductCategoryName] = "Bikes")
````

## FILTER
**Purpose**: Returns a table that represents a filtered version of another table. It is often used as a filter argument within CALCULATE.
**Syntax**:
````dax
FILTER(<table>, <filter_expression>)
````
**Example**: Calculate sales for orders placed in 2023.
````dax
Sales 2023 = CALCULATE(SUM(FactInternetSales[SalesAmount]), FILTER(Sales, YEAR(FactInternetSales[OrderDate]) = 2023))
````

## ALL
**Purpose**: Returns all the rows in a table, or all the values in a column, ignoring any filters that might have been applied.
**Syntax**:
````dax
ALL(<table>_or_<column>)
````
**Example**: Calculate percentage of total sales for a product.
````dax
% of Total Sales = DIVIDE(SUM(FactInternetSales[SalesAmount]), CALCULATE(SUM(FactInternetSales[SalesAmount]), ALL(Sales)))
````

## ALLEXCEPT
**Purpose**: Removes all context filters in the table except filters that have been applied to the specified columns.
**Syntax**:
````dax
ALLEXCEPT(<table>, <column1>, [<column2>]...)
````
**Example**: Calculate total sales for the current customer, ignoring other filters except for 'CustomerID'.
````dax
Customer Total Sales = CALCULATE(SUM(FactInternetSales[SalesAmount]), ALLEXCEPT(FactInternetSales, FactInternetSales[CustomerKey]))
````

## RELATED
**Purpose**: Returns a related value from another table. It requires an active relationship between the tables.
**Syntax**:
````dax
RELATED(<column>)
````
**Example**: Get the product name for each sale.
````dax
Product Category = RELATED('DimProductCategory'[EnglishProductCategoryName])
````

# 7. Table Manipulation Functions

These functions return a table, which can then be used by other DAX functions that require a table as input.

## ADDCOLUMNS
**Purpose**: Adds calculated columns to a given table.
**Syntax**:
````dax
ADDCOLUMNS(<table>, <name>, <expression>[, <name>, <expression>]...)
````
**Example**: Add a 'Profit' column to the Sales table.
````dax
Sales with Profit = ADDCOLUMNS(Sales, "Profit", Sales[SalesAmount] - Sales[Cost])
````

## SUMMARIZE
**Purpose**: Returns a summary table for the requested totals over a set of groups.
**Syntax**:
````dax
SUMMARIZE(<table>, <groupBy_columnName1>, [<groupBy_columnName2>]...[, <name>, <expression>]...)
````
**Example**: Summarize total sales by customer and product.
````dax
Sales Summary = SUMMARIZE(Sales, Customers[CustomerName], Products[ProductName],
                        "Total Sales", SUM(Sales[SalesAmount]))
````

## VALUES
**Purpose**: Returns a one-column table that contains the distinct values from the specified column.
**Syntax**:
````dax
VALUES(<column>)
````
**Example**: Get a list of unique product categories.
````dax
Unique Categories = VALUES(Products[Category])
````

## DISTINCT
**Purpose**: Returns a one-column table that contains the distinct values from the specified column. Similar to VALUES but handles blank differently.
**Syntax**:
````dax
DISTINCT(<column>)
````
**Example**: Get a list of unique product brands.
````dax
Unique Brands = DISTINCT(Products[Brand])
````

# 8. Time Intelligence Functions

These functions allow you to perform calculations across time periods (e.g., year-to-date, previous month). They require a well-formed date table.

## TOTALYTD
**Purpose**: Calculates the year-to-date total of an expression.
**Syntax**:
````dax
TOTALYTD(<expression>, <dates>[, <filter>][, <year_end_date>])
````
**Example**: Calculate year-to-date sales.
````dax
YTD Sales = TOTALYTD(SUM(Sales[SalesAmount]), 'Date'[Date])
````

## TOTALMTD
**Purpose**: Calculates the month-to-date total of an expression.
**Syntax**:
````dax
TOTALMTD(<expression>, <dates>[, <filter>])
````
**Example**: Calculate month-to-date sales.
````dax
MTD Sales = TOTALMTD(SUM(Sales[SalesAmount]), 'Date'[Date])
````

## TOTALQTD
**Purpose**: Calculates the quarter-to-date total of an expression.
**Syntax**:
````dax
TOTALQTD(<expression>, <dates>[, <filter>])
````
**Example**: Calculate quarter-to-date sales.
````dax
QTD Sales = TOTALQTD(SUM(Sales[SalesAmount]), 'Date'[Date])
````

## DATEADD
**Purpose**: Returns a table that contains a column of dates, shifted either forward or backward in time by a specified number of intervals from the dates in the current context.
**Syntax**:
````dax
DATEADD(<dates>, <number_of_intervals>, <interval>)
````
**Example**: Calculate sales from the previous month.
````dax
Sales Previous Month = CALCULATE(SUM(Sales[SalesAmount]), DATEADD('Date'[Date], -1, MONTH)))
````

## SAMEPERIODLASTYEAR
**Purpose**: Returns a table that contains a column of dates that are one year back in time from the dates in the specified dates column.
**Syntax**:
````dax
SAMEPERIODLASTYEAR(<dates>)
````
**Example**: Calculate sales from the same period last year.
````dax
Sales Last Year = CALCULATE(SUM(Sales[SalesAmount]), SAMEPERIODLASTYEAR('Date'[Date]))
````
