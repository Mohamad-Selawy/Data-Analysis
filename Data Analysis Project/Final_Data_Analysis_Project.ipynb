### Project Goal

This project is designed to provide hands-on experience with a typical data analyst workflow, integrating three essential tools: SQL Server, Python, and Power BI. ### Project Story: Junior Data Analyst at AdventureWorks

As a Junior Data Analyst, your task is to develop a new reporting dataset for the Sales Department. The company's standard practice avoids direct Power BI connections to large fact tables. Instead, you are required to implement a staged reporting pipeline:

1.  **Extract Data**: Read raw sales data from SQL Server.
2.  **Transform Data**: Clean and prepare the data using Python, adding relevant business features.
3.  **Load Data**: Save the refined dataset back into a dedicated reporting table in SQL Server.
4.  **Report**: Create a Power BI report based on this new reporting table.

This workflow simulates a real-world business scenario, demonstrating the end-to-end process of data analysis.

### Python Database Connection Code

Students will use the following exact Python code for connecting to SQL Server and interacting with data. This code should be used as provided, without modification, for database connectivity.

```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Setup Connection (Change server and database if needed)
engine = create_engine(
    "mssql+pyodbc://@localhost/AdventureWorksDW2022?driver=ODBC+Driver+17+for+SQL+Server"
)

# 2. Read data from SQL View into Pandas DataFrame
df = pd.read_sql_table("FactInternetSales", con=engine)
print("Data successfully read from SQL!")

# 3. Write data back to a new SQL Table
df.to_sql(
    name="PythonSalesReportingTable",
    con=engine,
    if_exists="replace",
    index=False,
)
print("Data successfully written to SQL table!")
```
