```python
# Check the data type of the Sales and Profit columns
print("\nData types for Sales and Profit before conversion attempt:")
print(df_orders[['Sales', 'Profit']].dtypes)

# Identify non-numeric entries in 'Sales' and 'Profit'
# Convert to numeric, coercing errors to NaN
df_orders['Sales_numeric'] = pd.to_numeric(df_orders['Sales'], errors='coerce')
df_orders['Profit_numeric'] = pd.to_numeric(df_orders['Profit'], errors='coerce')

# Find rows where conversion resulted in NaN, meaning they contained non-numeric characters
non_numeric_sales_entries = df_orders[df_orders['Sales_numeric'].isnull() & df_orders['Sales'].notnull()]['Sales']
non_numeric_profit_entries = df_orders[df_orders['Profit_numeric'].isnull() & df_orders['Profit'].notnull()][['Profit']]

display(non_numeric_sales_entries)
display(non_numeric_profit_entries)
```

```python
# Identify any records where Sales or Quantity are negative
negative_sales_records = df_orders[df_orders['Sales_numeric'] < 0]
negative_quantity_records = df_orders[df_orders['Quantity'] < 0]
negative_profit_records = df_orders[df_orders['Profit_numeric'] < 0]

print("\nRecords with Negative Sales:")
print(f"Found {len(negative_sales_records)} records with negative Sales values.")
display(negative_sales_records.head())

print("\nRecords with Negative Quantity:")
print(f"Found {len(negative_quantity_records)} records with negative Quantity values.")
display(negative_quantity_records.head())

print("\nRecords with Negative Profit (for context, already noted in 1.2):")
print(f"Found {len(negative_profit_records)} records with negative Profit values.")
display(negative_profit_records.head())
```

```python
# Identify any records where Sales appear to be unusually high
# Using a simple heuristic: Sales > 1.5 * Q3 for now, or a very high absolute value
Q1_sales = df_orders['Sales_numeric'].quantile(0.25)
Q3_sales = df_orders['Sales_numeric'].quantile(0.75)
IQR_sales = Q3_sales - Q1_sales
upper_bound_sales = Q3_sales + 1.5 * IQR_sales

unusually_high_sales_records = df_orders[df_orders['Sales_numeric'] > upper_bound_sales]

print(f"\nRecords with Unusually High Sales (Sales > {upper_bound_sales:.2f}):")
print(f"Found {len(unusually_high_sales_records)} records with unusually high Sales values.")
display(unusually_high_sales_records.sort_values(by='Sales_numeric', ascending=True).head())
```

### Business Scenario
Outliers, or data points that significantly deviate from other observations, can skew statistical analyses and provide misleading insights. ### Task 6.1: Identify and Characterize Outliers using the IQR Method
Apply the Interquartile Range (IQR) method to identify outliers in key numerical columns. The IQR method is robust to extreme values and is a commonly used technique for outlier detection. You will identify outliers and discuss their potential business implications without immediately removing them.

**Your Goals:**
*   For the `Sales` column, calculate the First Quartile (Q1), Third Quartile (Q3), and the Interquartile Range (IQR).
*   Determine the lower bound and upper bound for outlier detection in `Sales`.
*   Identify and count the number of outliers (both low and high) in the `Sales` column.
*   Repeat the process for the `Profit` column.
*   Display descriptive statistics (mean, median, standard deviation) for `Sales` and `Profit` before any outlier handling.

**Analytical Questions:**
1.  How do the calculated lower and upper bounds for `Sales` and `Profit` define what constitutes an outlier in this dataset?
2.  What is the count of high `Sales` outliers? What are the potential business reasons for such extremely high sales values (e.g., large corporate contracts, bulk purchases)?
3.  What is the count of low `Profit` outliers (or significant negative profit values)? What business scenarios could lead to such substantial losses on a single transaction?


```python
# --- Outlier Detection for 'Sales' ---

# Calculate Q1, Q3, and IQR for Sales
Q1_sales = df_orders['Sales'].quantile(0.25)
Q3_sales = df_orders['Sales'].quantile(0.75)
IQR_sales = Q3_sales - Q1_sales

# Determine lower and upper bounds for Sales outliers
lower_bound_sales = Q1_sales - 1.5 * IQR_sales
upper_bound_sales = Q3_sales + 1.5 * IQR_sales

# Identify and count Sales outliers
sales_outliers = df_orders[(df_orders['Sales'] < lower_bound_sales) | (df_orders['Sales'] > upper_bound_sales)]
print(f"\nSales Outlier Detection (IQR Method):\n")
print(f"Q1 (Sales): {Q1_sales:.2f}")
print(f"Q3 (Sales): {Q3_sales:.2f}")
print(f"IQR (Sales): {IQR_sales:.2f}")
print(f"Lower Bound (Sales): {lower_bound_sales:.2f}")
print(f"Upper Bound (Sales): {upper_bound_sales:.2f}")
print(f"Number of Sales Outliers: {len(sales_outliers)}")

# Display some Sales outliers for inspection
if not sales_outliers.empty:
    print("\nSample of Sales Outliers (first 5 rows):")
    display(sales_outliers.sort_values(by='Sales', ascending=False).head())
```

```python
# --- Outlier Detection for 'Profit' ---

# Calculate Q1, Q3, and IQR for Profit
Q1_profit = df_orders['Profit'].quantile(0.25)
Q3_profit = df_orders['Profit'].quantile(0.75)
IQR_profit = Q3_profit - Q1_profit

# Determine lower and upper bounds for Profit outliers
lower_bound_profit = Q1_profit - 1.5 * IQR_profit
upper_bound_profit = Q3_profit + 1.5 * IQR_profit

# Identify and count Profit outliers
profit_outliers = df_orders[(df_orders['Profit'] < lower_bound_profit) | (df_orders['Profit'] > upper_bound_profit)]
print(f"\nProfit Outlier Detection (IQR Method):\n")
print(f"Q1 (Profit): {Q1_profit:.2f}")
print(f"Q3 (Profit): {Q3_profit:.2f}")
print(f"IQR (Profit): {IQR_profit:.2f}")
print(f"Lower Bound (Profit): {lower_bound_profit:.2f}")
print(f"Upper Bound (Profit): {upper_bound_profit:.2f}")
print(f"Number of Profit Outliers: {len(profit_outliers)}")

# Display some Profit outliers for inspection
if not profit_outliers.empty:
    print("\nSample of Profit Outliers (first 5 rows):")
    display(profit_outliers.sort_values(by='Profit', ascending=False).head())
```
