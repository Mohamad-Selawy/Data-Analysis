## ✂️ 4. STEP 2: MISSION PREPARATION — PACK LIGHT (Subset Selection & Column Pruning)

**Golden Rule of Data Wrangling: Efficiency!** ⚡ As a seasoned detective, you never carry unnecessary luggage. We are **ONLY** interested in solving mysteries for the **year 2015**. Any other years are just background noise.

### The Pruning Rule:

Why clean data you'll never use? Identify which columns are strictly required to answer the 5 final business questions. Filter for orders from **2015 only**, keep only those essential columns, and discard the rest. Save your memory and computing power!

**Goal:** Create a new DataFrame containing *only* 2015 data and *only* the columns you deem necessary for the upcoming analysis.

---


```python
# First, ensure 'Order Date' is in datetime format to filter by year
df['Order Date'] = df['Order Date'].astype('datetime64[ns]')

# Filter for orders from the year 2015
df_2015 = df[df['Order Date'].dt.year == 2015].copy()

# Define essential columns for the 5 business questions
essential_columns = [
    'Row ID', 'Order ID', 'Order Date', 'Ship Date', 'Ship Mode',
    'Segment', 'State', 'Sales', 'Profit', 'Category', 'Discount'
]

# Select only the essential columns for df_2015
df_2015 = df_2015[essential_columns]

print(f"Original DataFrame shape: {df.shape}")
print(f"2015 DataFrame shape (after filtering and column pruning): {df_2015.shape}")
display(df_2015.head())
```

## 🧹 5. STEP 3: THE DEEP CLEANSE (Data Cleaning Pipeline)

Now for the gritty work, Detective! It's time for **The Deep Cleanse**. Using your pruned 2015 subset, resolve all the anomalies you logged in your Detective's Notebook from Step 1.

**Your Cleaning Checklist:**

*   Strip any sneaky leading/trailing spaces from string columns.
*   Fix inconsistent capitalization where appropriate.
*   Convert date columns to proper datetime formats.
*   Cast numerical columns (like 'Sales' or 'Profit') to appropriate numeric types (e.g., floats).
*   Strategically tackle any remaining null values (e.g., fill with a sensible value, or drop if necessary).
*   Vaporize any duplicate rows that survived the pruning.

Ensure your data is pristine and ready for analysis!

---


```python
# 1. Drop duplicate rows
initial_rows = df_2015.shape[0]
df_2015.drop_duplicates(inplace=True)
print(f"Dropped {initial_rows - df_2015.shape[0]} duplicate rows.")

# 2. Convert 'Ship Date' to datetime (Order Date was already converted in the previous step)
df_2015['Ship Date'] = df_2015['Ship Date'].astype('datetime64[ns]')

# 3. Clean and convert 'Sales' and 'Profit' to numeric
# Sales might contain non-numeric characters like '$' or commas, so clean first
df_2015['Sales'] = df_2015['Sales'].astype(str).str.replace('$', '').str.replace(',', '')
df_2015['Sales'] = df_2015['Sales'].astype('float64')
df_2015['Profit'] = df_2015['Profit'].astype('float64')

# Fill missing values for Sales and Profit with 0, as missing values likely mean no sale/profit.
df_2015['Sales'].fillna(0, inplace=True)
df_2015['Profit'].fillna(0, inplace=True)

# 4. Fill missing values for categorical columns 'Ship Mode' and 'Category'
df_2015['Ship Mode'].fillna('Unknown', inplace=True)
df_2015['Category'].fillna('Unknown', inplace=True)

# 5. Strip whitespace and normalize casing for string columns
string_cols = ['Ship Mode', 'Segment', 'State', 'Category']
for col in string_cols:
    if col in df_2015.columns:
        df_2015[col] = df_2015[col].astype(str).str.strip().str.title()

print("\n--- Cleaning Summary ---")
print("Data types after cleaning:")
display(df_2015.info())
print("\nMissing values after cleaning:")
display(df_2015.isnull().sum())
print("\nFirst 5 rows of cleaned data:")
display(df_2015.head())
```

### 📊 Question 1 (The Baseline): What are the total sales and total profit generated by the Superstore in 2015?


```python
# Calculate total sales and total profit for 2015
total_sales_2015 = df_2015['Sales'].sum()
total_profit_2015 = df_2015['Profit'].sum()

print(f"Total Sales in 2015: ${total_sales_2015}")
print(f"Total Profit in 2015: ${total_profit_2015}")
```

### 👥 Question 2 (The Target Market): Which customer segment (Consumer, Corporate, or Home Office) was the most profitable in 2015?


```python
# Calculate profit by customer segment
profit_by_segment = df_2015.groupby('Segment')['Profit'].sum().sort_values(ascending=False)

print("Profit by Customer Segment in 2015:")
display(profit_by_segment)
```

#### 💡 Detective's Business Insight:
In 2015, the **Consumer** segment was the most profitable, generating **$27,783.72** in profit. This indicates that individual customers are a key driver of the Superstore's bottom line, followed by Corporate and then Home Office segments. This insight can help in tailoring marketing strategies and resource allocation to further nurture the most profitable customer base.

### 🗺️ Question 3 (The Hotspots): What are the top 5 states by total Sales in 2015, and what were their corresponding profits?


```python
# Calculate total sales and profit by state, then sort by sales
state_performance = df_2015.groupby('State').agg(
    {'Sales': 'sum',
    'Profit': 'sum'}
).sort_values(by = 'Sales', ascending=False)

# Get the top 5 states by sales
top_5_states = state_performance.head(5)

print("Top 5 States by Total Sales in 2015:")
display(top_5_states)
```

#### 💡 Detective's Business Insight:
The top 5 states by total sales in 2015 were New York, Washington, Ohio, Georgia, and Missouri. While New York leads significantly in both sales and profit, Ohio stands out as a state with high sales but a negative profit, indicating potential issues with pricing, discounts, or operational costs in that region. This suggests a need for further investigation into Ohio's market strategy.

### 🚚 Question 4 (The Logistics Race): Which shipping mode was most frequently utilized in 2015, and what was its average delivery time (Ship Date minus Order Date) in days?


```python
# Calculate the delivery time for each order
df_2015['Delivery_Time'] = (df_2015['Ship Date'] - df_2015['Order Date']).dt.days

# Find the most frequently utilized shipping mode
most_frequent_ship_mode = df_2015['Ship Mode'].mode()[0]

# Calculate the average delivery time for the most frequent shipping mode
average_delivery_time = df_2015[df_2015['Ship Mode'] == most_frequent_ship_mode]['Delivery_Time'].mean()

print(f"Most Frequently Utilized Ship Mode: {most_frequent_ship_mode}")
print(f"Average Delivery Time for '{most_frequent_ship_mode}': {average_delivery_time:.2f} days")
```

#### 💡 Detective's Business Insight:
In 2015, **Standard Class** was the most frequently utilized shipping mode, with an average delivery time of approximately **4.97 days**. This highlights its critical role in the Superstore's logistics. While efficient, understanding this benchmark allows for future optimization and comparison against other shipping options, ensuring competitive and satisfactory delivery experiences.

### 🎟️ Question 5 (The Discount Trap): What are the top product categories by sales in 2015, and what was the average discount offered for each?


```python
# Calculate total sales and average discount by product category
category_performance = df_2015.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Average_Discount=('Discount', 'mean')
).sort_values(by='Total_Sales', ascending=False)

print("Top Product Categories by Sales in 2015 and their Average Discounts:")
display(category_performance)
```

#### 💡 Detective's Business Insight:
In 2015, **Office Supplies** led significantly in sales with **6,117,944**, followed by **Technology** with **$1,140,988**. The average discounts for these categories were 16.20% and 12.71% respectively. The presence of 'Offcie Supplies' and 'Furnitures' suggests further normalization of categorical data could be beneficial. This analysis highlights the importance of Office Supplies to the Superstore's revenue, but also raises questions about the profitability of discounts within this category, especially given the presence of similar categories with slightly different spellings and the 'Unknown' category.
