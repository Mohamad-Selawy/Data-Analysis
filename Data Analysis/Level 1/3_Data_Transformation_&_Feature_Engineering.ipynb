## 2. Sorting & Indexing

Sorting data allows us to arrange DataFrame rows based on the values in one or more columns or by the DataFrame's index. This is fundamental for organizing and analyzing data.

### 2.1 Sort by Column Values (`.sort_values()`)

Use `.sort_values()` to sort the DataFrame by the values of a specified column. You can sort in ascending (default) or descending order. Here, we'll sort the DataFrame by `Price` in descending order.

```python
df_sorted_salary = df.sort_values(by='Salary', ascending=False)
display(df_sorted_salary.head())
```

### 2.2 Sort by Index (`.sort_index()`)

The `.sort_index()` method sorts the DataFrame by its index labels. First, we'll set `OrderID` as the index, then sort by it. If no index is set, it sorts by the default integer index.

```python
df_indexed = df.set_index('EmployeeID').sort_index()
display(df_indexed.head())
```

## 3. Feature Creation & Vectorization

Feature creation involves deriving new columns from existing ones. Vectorized operations and conditional logic are powerful tools to achieve this efficiently.

### 3.1 Deriving New Columns with Vectorized Operations

Vectorized operations apply an operation to an entire array or series at once, which is significantly faster than row-by-row operations. Here, we'll calculate an `Effective_Salary` by multiplying `Salary` with `PerformanceScore` (normalized).

```python
df['Effective_Salary'] = df['Salary'] * (df['PerformanceScore'] / 100)
display(df.head())
```

### 3.2 Conditional Feature Creation



```python
avg_effective_salary = df['Effective_Salary'].mean()
df['Performance_Segment'] = df['Effective_Salary'].apply(lambda x: 'High Performer' if x > avg_effective_salary else 'Standard Performer')
display(df.head())
```

## 4. Applying Functions: `.apply()`, `.map()`, and `lambda`

These are versatile tools for applying custom logic or transformations to DataFrame Series or elements.

### 4.1 Using `.apply()` with `lambda` for Conditional Logic

`.apply()` can be used with a `lambda` function to define custom logic for creating new columns based on conditions or transformations. Here, we'll categorize products based on whether their `Price` is above or below the average price.

```python
avg_salary = df['Salary'].mean()
df['Salary_Tier'] = df['Salary'].apply(lambda x: 'High' if x > avg_salary else 'Low')
display(df.head())
```

### 4.2 Using `.map()` for Value Substitution

`.map()` is highly efficient for substituting values in a Series based on a dictionary or another Series. It's ideal for one-to-one mappings. We'll map `Category` to broader `Category_Group` names.

```python
department_mapping = {
    'IT': 'Tech Dept',
    'HR': 'Admin Dept',
    'Finance': 'Finance Dept',
    'Marketing': 'Sales & Marketing',
    'Sales': 'Sales & Marketing'
}
df['Department_Group'] = df['Department'].map(department_mapping).fillna('Other')
display(df.head())
```

## 5. Merging, Joining, and Concatenating

Merging, joining, and concatenating are essential operations for combining DataFrames, allowing you to integrate information from different sources based on common columns or by simply stacking them.

### 5.1 Merging DataFrames (`pd.merge()`)

`pd.merge()` is used to combine two DataFrames based on common columns (keys). It's similar to SQL joins and offers various types (inner, outer, left, right). Below, we'll demonstrate an inner merge.

```python
# Create df_employees_subset from the main df
df_employees_subset = df[['EmployeeID', 'Name', 'Department']].head(5)

# Create a sample DataFrame with additional contact information for demonstration
df_contact_info = pd.DataFrame({
    'EmployeeID': [101, 103, 105, 106], # Include some IDs from df_employees_subset and some new ones
    'Email': ['alice.s@example.com', 'charlie.b@example.com', 'eve.a@example.com', 'frank.w@example.com'],
    'Phone': ['555-1001', '555-1003', '555-1005', '555-1006']
})

print("df_employees_subset (derived from main df):")
display(df_employees_subset)
print("\ndf_contact_info:")
display(df_contact_info)

# Perform an inner merge on 'EmployeeID'
# This will show only employees present in both DataFrames
merged_inner_df = pd.merge(df_employees_subset, df_contact_info, on='EmployeeID', how='inner')
print("\nInner Merged DataFrame (employees with contact info):")
display(merged_inner_df)

# Also demonstrate a left merge to include all employees and their contact info if available
merged_left_df = pd.merge(df_employees_subset, df_contact_info, on='EmployeeID', how='left')
print("\nLeft Merged DataFrame (all employees, with contact info where available):")
display(merged_left_df)
```

### 5.2 Concatenating DataFrames (`pd.concat()`)

`pd.concat()` stacks DataFrames either vertically (row-wise) or horizontally (column-wise) along an axis. It's used to append or stack multiple DataFrames directly.

```python
# Create two sample DataFrames for demonstration
df_part1 = pd.DataFrame({
    'Name': ['Frank', 'Grace'],
    'Age': [30, 25]
})

df_part2 = pd.DataFrame({
    'Name': ['Harry', 'Ivy'],
    'Age': [35, 28]
})

print("df_part1:")
display(df_part1)
print("\ndf_part2:")
display(df_part2)

# Concatenate them vertically
concatenated_df = pd.concat([df_part1, df_part2], ignore_index=True)
print("\nVertically Concatenated DataFrame:")
display(concatenated_df)
```
