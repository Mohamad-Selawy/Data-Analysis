# GroupBy and Pivot Tables: Advanced Data Summarization

_Use `groupby()` to split-apply-combine data, and `pivot_table()` to reshape and summarize_

## What You'll Learn
- How to use `groupby()` to compute grouped statistics.
- How to create pivot tables for aggregated summaries using `pd.pivot_table()`.

## Setup: Import Libraries and Load Data

```python
import pandas as pd
```

## Load Dataset

We will load the `Data_Selection_&_Filtering.csv` dataset into a pandas DataFrame (`df`). This dataset will be used throughout the notebook to demonstrate `groupby()`, `pivot_table()`, and `df.pivot()` functionalities.

```python
# Load the dataset
df = pd.read_csv('Data_Selection_&_Filtering.csv')

print("First 5 rows of the dataset:")
display(df.head())
```

## 1. `groupby()` - Split, Apply, Combine

The `groupby()` method is fundamental for aggregating data. It involves three steps: **Split** (dividing data into groups based on some criteria), **Apply** (applying a function to each group independently), and **Combine** (combining the results into a data structure). This section demonstrates grouping by single and multiple columns, and applying multiple aggregation functions.

---
### Grouping by a Single Column

**Business Question:** *What is the average `Salary` for each `Department`?*

We will group the data by `Department` and calculate the mean of the `Salary` column.

```python
# Group by 'Department' and calculate the mean of 'Salary'
avg_salary_by_department = df.groupby('Department')['Salary'].mean()
display(avg_salary_by_department)
```

---
### Grouping by Multiple Columns

**Business Question:** *What is the total `PerformanceScore` for each `Department` within each `City`?*

Here, we group by both `Department` and `City` to get a more granular breakdown.

```python
# Group by 'Department' and 'City', then sum 'PerformanceScore'
total_performance_by_dept_city = df.groupby(['Department', 'City'])['PerformanceScore'].sum()
display(total_performance_by_dept_city)
```

---
### Applying Multiple Aggregation Functions

**Business Question:** *For each `Department`, what are the average, maximum, and minimum `Salary`, and the median `Experience`?*

The `.agg()` method allows us to apply different aggregation functions to one or more columns simultaneously.


```python
# Using dictionary syntax for .agg()
dept_stats = df.groupby('Department').agg({
    'Salary': ['mean', 'max', 'min'],
    'Experience': 'median',
    'PerformanceScore': 'mean'
})
display(dept_stats)
```

## 2. `pivot_table()` - Flexible Table Generator

The `pivot_table()` function is used to create spreadsheet-style pivot tables as a DataFrame. It summarizes data by specifying `index`, `columns`, `values`, and `aggfunc`.

---
### Basic Pivot Table

**Business Question:** *How does the average `Salary` compare across different `Department` by `City`?*

We will create a pivot table with `Department` as rows (index), `City` as columns, and the average `Salary` as the values.

```python
# Pivot table showing average Salary by Department and City
avg_salary_pivot = pd.pivot_table(df,
                                       values='Salary',
                                       index='Department',
                                       columns='City',
                                       aggfunc='mean')
display(avg_salary_pivot)
```

```python
# Pivot table with margins
# margins=True adds row and column totals ('All')
pivot_with_totals = pd.pivot_table(df,
                                         values='Salary',
                                         index='Department',
                                         columns='City',
                                         aggfunc='mean',
                                         margins=True) # Adds row and column totals
display(pivot_with_totals)
```

---
### Practice Exercise: Pivot Table

**Challenge:** *Calculate the total `PerformanceScore` for each `Gender` and `City` combination.*

**Hint:** Use `pd.pivot_table()` with `index`, `columns`, `values`, and `aggfunc`.

```python
# Pivot table showing total PerformanceScore by Gender and City
pivot_performance_by_gender_city = pd.pivot_table(df,
                                              values='PerformanceScore',
                                              index='Gender',
                                              columns='City',
                                              aggfunc='sum')
display(pivot_performance_by_gender_city)
```

### Summary: GroupBy & Pivot Tables

*   `groupby()` is excellent for programmatic analysis and applying various aggregations (including custom functions) on grouped data.
*   `pivot_table()` is ideal for creating cross-tabulations or spreadsheet-style summaries for reporting and multi-dimensional analysis.
